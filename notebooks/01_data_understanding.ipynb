````python
# Step 8: Overall Profit Margin
# This cell independently loads the dataset if df is not available.

from pathlib import Path
import pandas as pd

# Find the project root
BASE_DIR = Path.cwd().parent

# Find Nassau Candy CSV file
csv_files = list((BASE_DIR / "data").glob("*.csv"))

nassau_files = [
    file for file in csv_files
    if "Nassau Candy Distributor" in file.name
]

if len(nassau_files) == 0:
    raise FileNotFoundError(
        "Nassau Candy Distributor CSV file was not found in the data folder."
    )

# Load dataset
DATA_PATH = nassau_files[0]
df = pd.read_csv(DATA_PATH)

# Calculate totals directly from the dataset
total_sales = df["Sales"].sum()
total_profit = df["Gross Profit"].sum()

# Calculate profit margin safely
if total_sales > 0:
    profit_margin = (total_profit / total_sales) * 100
else:
    profit_margin = 0

print("========== PROFIT ANALYSIS ==========")
print(f"Dataset loaded : {DATA_PATH.name}")
print(f"Rows           : {len(df):,}")
print(f"Total Sales    : ${total_sales:,.2f}")
print(f"Total Profit   : ${total_profit:,.2f}")
print(f"Profit Margin  : {profit_margin:.2f}%")
````

````python
# Step 9: Top 10 Products by Sales

# Create product-level analysis
product_analysis = (
    df.groupby(["Product ID", "Product Name"], as_index=False)
      .agg(
          Total_Sales=("Sales", "sum"),
          Total_Units=("Units", "sum"),
          Total_Gross_Profit=("Gross Profit", "sum"),
          Total_Cost=("Cost", "sum")
      )
      .sort_values("Total_Sales", ascending=False)
)

# Display Top 10 Products
print("========== TOP 10 PRODUCTS BY SALES ==========")

top_10_products = product_analysis.head(10).copy()

print(top_10_products.to_string(index=False))

print("\nTop product by sales:")
print(top_10_products.iloc[0]["Product Name"])

print(f"\nHighest product sales: ${top_10_products.iloc[0]['Total_Sales']:,.2f}")
````

````python
# Step 10: Product Profitability Analysis

# Calculate profit margin for each product
product_analysis["Profit_Margin_%"] = (
    product_analysis["Total_Gross_Profit"] /
    product_analysis["Total_Sales"]
) * 100

# Handle any invalid values
product_analysis["Profit_Margin_%"] = (
    product_analysis["Profit_Margin_%"]
    .replace([float("inf"), -float("inf")], 0)
    .fillna(0)
)

# Sort products by profit margin
profitability_analysis = product_analysis.sort_values(
    "Profit_Margin_%",
    ascending=False
).copy()

print("========== TOP 10 PRODUCTS BY PROFIT MARGIN ==========")

top_10_profitable = profitability_analysis.head(10)

print(
    top_10_profitable[
        [
            "Product ID",
            "Product Name",
            "Total_Sales",
            "Total_Units",
            "Total_Gross_Profit",
            "Total_Cost",
            "Profit_Margin_%"
        ]
    ].to_string(index=False)
)

print("\n========== MOST PROFITABLE PRODUCT ==========")

best_product = top_10_profitable.iloc[0]

print(f"Product Name      : {best_product['Product Name']}")
print(f"Total Sales       : ${best_product['Total_Sales']:,.2f}")
print(f"Gross Profit      : ${best_product['Total_Gross_Profit']:,.2f}")
print(f"Profit Margin     : {best_product['Profit_Margin_%']:.2f}%")
````

````python
# ============================================================
# STEP 11: SHIPPING PERFORMANCE ANALYSIS
# ============================================================

import pandas as pd

# ------------------------------------------------------------
# 1. Make sure both date columns are datetime
# ------------------------------------------------------------

df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    errors="coerce"
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"],
    errors="coerce"
)

# ------------------------------------------------------------
# 2. Check whether any invalid dates were created
# ------------------------------------------------------------

invalid_order_dates = df["Order Date"].isna().sum()
invalid_ship_dates = df["Ship Date"].isna().sum()

print("Invalid Order Dates :", invalid_order_dates)
print("Invalid Ship Dates  :", invalid_ship_dates)

# ------------------------------------------------------------
# 3. Calculate delivery time in days
# ------------------------------------------------------------

df["Delivery_Days"] = (
    df["Ship Date"] - df["Order Date"]
).dt.days

# ------------------------------------------------------------
# 4. Check invalid delivery values
# ------------------------------------------------------------

invalid_delivery = (df["Delivery_Days"] < 0).sum()

print("Invalid Delivery Days:", invalid_delivery)

# ------------------------------------------------------------
# 5. Shipping Performance Summary
# ------------------------------------------------------------

print("\n========== SHIPPING PERFORMANCE ==========")

print(
    f"Average Delivery Days : "
    f"{df['Delivery_Days'].mean():.2f}"
)

print(
    f"Minimum Delivery Days : "
    f"{df['Delivery_Days'].min():.0f}"
)

print(
    f"Maximum Delivery Days : "
    f"{df['Delivery_Days'].max():.0f}"
)

print(
    f"Median Delivery Days  : "
    f"{df['Delivery_Days'].median():.2f}"
)

# ------------------------------------------------------------
# 6. Shipping mode performance
# ------------------------------------------------------------

shipping_analysis = (
    df.groupby("Ship Mode")
      .agg(
          Orders=("Order ID", "count"),
          Average_Delivery_Days=("Delivery_Days", "mean"),
          Minimum_Delivery_Days=("Delivery_Days", "min"),
          Maximum_Delivery_Days=("Delivery_Days", "max")
      )
      .sort_values("Average_Delivery_Days")
)

print("\n========== SHIPPING MODE ANALYSIS ==========")

print(shipping_analysis.round(2).to_string())
````

````python
# ============================================================
# STEP 12: SHIPPING MODE PERFORMANCE ANALYSIS
# ============================================================

# Analyze shipping performance by Ship Mode

shipping_mode_analysis = (
    df.groupby("Ship Mode")
      .agg(
          Total_Orders=("Order ID", "count"),
          Total_Sales=("Sales", "sum"),
          Total_Units=("Units", "sum"),
          Total_Gross_Profit=("Gross Profit", "sum"),
          Average_Delivery_Days=("Delivery_Days", "mean")
      )
      .reset_index()
)

# Calculate profit margin for each shipping mode
shipping_mode_analysis["Profit_Margin_%"] = (
    shipping_mode_analysis["Total_Gross_Profit"]
    / shipping_mode_analysis["Total_Sales"]
) * 100

# Sort by average delivery time
shipping_mode_analysis = shipping_mode_analysis.sort_values(
    "Average_Delivery_Days"
)

# Display results
print("========== SHIPPING MODE PERFORMANCE ==========\n")

print(
    shipping_mode_analysis.to_string(
        index=False,
        formatters={
            "Total_Sales": "${:,.2f}".format,
            "Total_Gross_Profit": "${:,.2f}".format,
            "Average_Delivery_Days": "{:.2f}".format,
            "Profit_Margin_%": "{:.2f}%".format
        }
    )
)

# ------------------------------------------------------------
# Best and slowest shipping modes
# ------------------------------------------------------------

fastest_mode = shipping_mode_analysis.iloc[0]["Ship Mode"]
fastest_delivery = shipping_mode_analysis.iloc[0]["Average_Delivery_Days"]

slowest_mode = shipping_mode_analysis.iloc[-1]["Ship Mode"]
slowest_delivery = shipping_mode_analysis.iloc[-1]["Average_Delivery_Days"]

print("\n========== SHIPPING INSIGHTS ==========")

print(
    f"Fastest Shipping Mode : {fastest_mode}"
)

print(
    f"Average Delivery Time : {fastest_delivery:.2f} days"
)

print(
    f"\nSlowest Shipping Mode : {slowest_mode}"
)

print(
    f"Average Delivery Time : {slowest_delivery:.2f} days"
)
````

````python
# ============================================================
# STEP 12B: LEAD TIME DATA QUALITY ANALYSIS
# ============================================================

print("========== LEAD TIME DATA QUALITY ANALYSIS ==========\n")

# Basic statistics
lead_time_stats = df["Delivery_Days"].describe()

print("Lead Time Statistics:")
print(lead_time_stats)

# Percentiles
q1 = df["Delivery_Days"].quantile(0.25)
median = df["Delivery_Days"].quantile(0.50)
q3 = df["Delivery_Days"].quantile(0.75)

print("\n========== LEAD TIME PERCENTILES ==========")

print(f"25th Percentile : {q1:.2f} days")
print(f"Median          : {median:.2f} days")
print(f"75th Percentile : {q3:.2f} days")

# IQR
iqr = q3 - q1

lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)

print("\n========== IQR ANALYSIS ==========")

print(f"IQR             : {iqr:.2f} days")
print(f"Lower Bound     : {lower_bound:.2f} days")
print(f"Upper Bound     : {upper_bound:.2f} days")

# Count potential outliers
outlier_count = (
    (df["Delivery_Days"] < lower_bound) |
    (df["Delivery_Days"] > upper_bound)
).sum()

print(f"Potential Outliers : {outlier_count:,}")

# Negative lead times
negative_count = (df["Delivery_Days"] < 0).sum()

print(f"Negative Lead Times : {negative_count:,}")

# Zero-day shipments
zero_day_count = (df["Delivery_Days"] == 0).sum()

print(f"Zero-Day Shipments : {zero_day_count:,}")

print("\n========== DATA QUALITY RESULT ==========")

if negative_count == 0:
    print("✓ No negative lead times found.")
else:
    print("⚠ Negative lead times detected.")

if outlier_count == 0:
    print("✓ No statistical lead-time outliers found.")
else:
    print(f"⚠ {outlier_count:,} potential lead-time outliers detected.")
````

````python
# ============================================================
# STEP 12D-FINAL: RELOAD RAW DATES & CALCULATE LEAD TIME
# ============================================================

print("========== FINAL DATE CORRECTION ==========\n")

# IMPORTANT:
# Reload ONLY the original date columns from the raw CSV.
# This recovers the original date strings that were converted
# to NaT in the previous cells.

raw_dates = pd.read_csv(
    "../data/Nassau Candy Distributor.csv",
    usecols=["Order Date", "Ship Date"]
)

# Correct date format: DD-MM-YYYY
raw_dates["Order Date"] = pd.to_datetime(
    raw_dates["Order Date"],
    format="%d-%m-%Y",
    errors="raise"
)

raw_dates["Ship Date"] = pd.to_datetime(
    raw_dates["Ship Date"],
    format="%d-%m-%Y",
    errors="raise"
)

# Replace the damaged date columns in df
df["Order Date"] = raw_dates["Order Date"]
df["Ship Date"] = raw_dates["Ship Date"]

# Calculate Lead / Delivery Time
df["Delivery_Days"] = (
    df["Ship Date"] - df["Order Date"]
).dt.days

# ============================================================
# VALIDATION
# ============================================================

print("Total Rows              :", len(df))
print("Missing Order Date      :", df["Order Date"].isna().sum())
print("Missing Ship Date       :", df["Ship Date"].isna().sum())
print("Missing Delivery Days   :", df["Delivery_Days"].isna().sum())

print("\n========== DATE RANGE ==========")

print("Order Date Minimum      :", df["Order Date"].min())
print("Order Date Maximum      :", df["Order Date"].max())

print("Ship Date Minimum       :", df["Ship Date"].min())
print("Ship Date Maximum       :", df["Ship Date"].max())

print("\n========== DELIVERY DAYS ==========")

print(df["Delivery_Days"].describe())

negative_count = (df["Delivery_Days"] < 0).sum()
zero_count = (df["Delivery_Days"] == 0).sum()

print("\nNegative Delivery Days  :", negative_count)
print("Zero-Day Shipments      :", zero_count)

print("\n========== FINAL RESULT ==========")

if (
    len(df) == 10194
    and df["Order Date"].isna().sum() == 0
    and df["Ship Date"].isna().sum() == 0
    and df["Delivery_Days"].isna().sum() == 0
    and negative_count == 0
):
    print("✓ SUCCESS")
    print("✓ All 10,194 records have valid dates.")
    print("✓ All 10,194 records have valid Delivery_Days.")
    print("✓ No negative delivery times found.")
else:
    print("⚠ VALIDATION FAILED — DO NOT CONTINUE")
````

````python
# ============================================================
# STEP 12E: LEAD TIME DISTRIBUTION & OPERATIONAL SEGMENTATION
# ============================================================

print("========== LEAD TIME DISTRIBUTION ANALYSIS ==========\n")

# Calculate data-driven thresholds
q1 = df["Delivery_Days"].quantile(0.25)
median = df["Delivery_Days"].quantile(0.50)
q3 = df["Delivery_Days"].quantile(0.75)

print("Lead Time Thresholds:")
print(f"Q1 / 25th Percentile  : {q1:.0f} days")
print(f"Median / 50th         : {median:.0f} days")
print(f"Q3 / 75th Percentile  : {q3:.0f} days")

# ------------------------------------------------------------
# Create operational lead-time segments
# ------------------------------------------------------------

def classify_lead_time(days):
    if days <= q1:
        return "Fast"
    elif days <= q3:
        return "Moderate"
    else:
        return "Slow"

df["Lead_Time_Category"] = df["Delivery_Days"].apply(
    classify_lead_time
)

# ------------------------------------------------------------
# Category counts
# ------------------------------------------------------------

category_counts = (
    df["Lead_Time_Category"]
    .value_counts()
    .reindex(["Fast", "Moderate", "Slow"])
)

category_percentage = (
    category_counts / len(df) * 100
)

print("\n========== LEAD TIME CATEGORIES ==========")

for category in ["Fast", "Moderate", "Slow"]:
    print(
        f"{category:<10} : "
        f"{category_counts[category]:,} orders "
        f"({category_percentage[category]:.2f}%)"
    )

# ------------------------------------------------------------
# Category performance
# ------------------------------------------------------------

category_analysis = (
    df.groupby("Lead_Time_Category", observed=True)
    .agg(
        Orders=("Order ID", "nunique"),
        Records=("Order ID", "size"),
        Average_Lead_Time=("Delivery_Days", "mean"),
        Median_Lead_Time=("Delivery_Days", "median"),
        Total_Sales=("Sales", "sum"),
        Total_Units=("Units", "sum"),
        Total_Gross_Profit=("Gross Profit", "sum")
    )
    .reindex(["Fast", "Moderate", "Slow"])
)

print("\n========== CATEGORY PERFORMANCE ==========")
print(category_analysis.to_string())

# ------------------------------------------------------------
# Final validation
# ------------------------------------------------------------

print("\n========== VALIDATION ==========")

print("Total records classified :", df["Lead_Time_Category"].notna().sum())
print("Unclassified records     :", df["Lead_Time_Category"].isna().sum())

if df["Lead_Time_Category"].isna().sum() == 0:
    print("✓ All records successfully classified.")
else:
    print("⚠ Some records were not classified.")

print("\n✓ STEP 12E COMPLETED")
````

````python
# ============================================================
# STEP 13: REGION & DIVISION LEAD-TIME ANALYSIS
# ============================================================

print("========== REGION & DIVISION LEAD-TIME ANALYSIS ==========\n")

# ------------------------------------------------------------
# 1. REGION ANALYSIS
# ------------------------------------------------------------

region_analysis = (
    df.groupby("Region", observed=True)
    .agg(
        Orders=("Order ID", "nunique"),
        Records=("Order ID", "size"),
        Average_Lead_Time=("Delivery_Days", "mean"),
        Median_Lead_Time=("Delivery_Days", "median"),
        Total_Sales=("Sales", "sum"),
        Total_Units=("Units", "sum"),
        Total_Gross_Profit=("Gross Profit", "sum")
    )
    .sort_values("Average_Lead_Time", ascending=False)
)

print("========== REGION PERFORMANCE ==========")
print(region_analysis.to_string())

# ------------------------------------------------------------
# 2. DIVISION ANALYSIS
# ------------------------------------------------------------

division_analysis = (
    df.groupby("Division", observed=True)
    .agg(
        Orders=("Order ID", "nunique"),
        Records=("Order ID", "size"),
        Average_Lead_Time=("Delivery_Days", "mean"),
        Median_Lead_Time=("Delivery_Days", "median"),
        Total_Sales=("Sales", "sum"),
        Total_Units=("Units", "sum"),
        Total_Gross_Profit=("Gross Profit", "sum")
    )
    .sort_values("Average_Lead_Time", ascending=False)
)

print("\n========== DIVISION PERFORMANCE ==========")
print(division_analysis.to_string())

# ------------------------------------------------------------
# 3. REGION × LEAD-TIME CATEGORY
# ------------------------------------------------------------

region_category = pd.crosstab(
    df["Region"],
    df["Lead_Time_Category"],
    normalize="index"
) * 100

region_category = region_category.reindex(
    columns=["Fast", "Moderate", "Slow"],
    fill_value=0
)

print("\n========== REGION × LEAD-TIME CATEGORY (%) ==========")
print(region_category.round(2).to_string())

# ------------------------------------------------------------
# 4. DIVISION × LEAD-TIME CATEGORY
# ------------------------------------------------------------

division_category = pd.crosstab(
    df["Division"],
    df["Lead_Time_Category"],
    normalize="index"
) * 100

division_category = division_category.reindex(
    columns=["Fast", "Moderate", "Slow"],
    fill_value=0
)

print("\n========== DIVISION × LEAD-TIME CATEGORY (%) ==========")
print(division_category.round(2).to_string())

# ------------------------------------------------------------
# 5. IDENTIFY BOTTLENECKS
# ------------------------------------------------------------

slowest_region = region_analysis["Average_Lead_Time"].idxmax()
fastest_region = region_analysis["Average_Lead_Time"].idxmin()

slowest_division = division_analysis["Average_Lead_Time"].idxmax()
fastest_division = division_analysis["Average_Lead_Time"].idxmin()

print("\n========== KEY FINDINGS ==========")

print(
    f"Slowest Region   : {slowest_region} "
    f"({region_analysis.loc[slowest_region, 'Average_Lead_Time']:.2f} days)"
)

print(
    f"Fastest Region   : {fastest_region} "
    f"({region_analysis.loc[fastest_region, 'Average_Lead_Time']:.2f} days)"
)

print(
    f"Slowest Division : {slowest_division} "
    f"({division_analysis.loc[slowest_division, 'Average_Lead_Time']:.2f} days)"
)

print(
    f"Fastest Division : {fastest_division} "
    f"({division_analysis.loc[fastest_division, 'Average_Lead_Time']:.2f} days)"
)

print("\n✓ STEP 13 COMPLETED")
````

````python
# ============================================================
# STEP 16: PRODUCT OPERATIONAL PRIORITY / RISK MATRIX
# ROBUST VERSION - SAFE AFTER KERNEL RESTART
# ============================================================

import pandas as pd
from pathlib import Path

print("========== STEP 16: PRODUCT OPERATIONAL PRIORITY ==========\n")


# ------------------------------------------------------------
# 1. ENSURE DATAFRAME EXISTS
# ------------------------------------------------------------

if "df" not in globals():

    print("df not found in current kernel.")
    print("Loading Nassau Candy dataset...\n")

    data_folder = Path("../data")

    csv_files = list(data_folder.glob("*.csv"))

    if not csv_files:
        raise FileNotFoundError(
            "No CSV file found inside the data folder."
        )

    # Find Nassau Candy CSV
    nassau_files = [
        f for f in csv_files
        if "nassau" in f.name.lower()
    ]

    if not nassau_files:
        raise FileNotFoundError(
            "Nassau Candy CSV file was not found."
        )

    csv_path = nassau_files[0]

    df = pd.read_csv(csv_path)

    print(f"Dataset loaded: {csv_path.name}")
    print(f"Rows loaded: {len(df):,}")


else:

    print("Existing dataframe 'df' found.")
    print(f"Rows available: {len(df):,}")


# ------------------------------------------------------------
# 2. ENSURE DATE COLUMNS ARE CORRECT
# ------------------------------------------------------------

df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    dayfirst=True,
    errors="coerce"
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"],
    dayfirst=True,
    errors="coerce"
)


# ------------------------------------------------------------
# 3. ENSURE DELIVERY DAYS EXISTS
# ------------------------------------------------------------

df["Delivery_Days"] = (
    df["Ship Date"] - df["Order Date"]
).dt.days


# ------------------------------------------------------------
# 4. VALIDATE DELIVERY DAYS
# ------------------------------------------------------------

missing_delivery = df["Delivery_Days"].isna().sum()
negative_delivery = (df["Delivery_Days"] < 0).sum()

print("\n========== DATA VALIDATION ==========")

print(f"Total Records       : {len(df):,}")
print(f"Missing Delivery    : {missing_delivery:,}")
print(f"Negative Delivery   : {negative_delivery:,}")


if missing_delivery > 0:
    raise ValueError(
        "Delivery_Days contains missing values. "
        "Fix date parsing before continuing."
    )

if negative_delivery > 0:
    raise ValueError(
        "Negative Delivery_Days detected. "
        "Fix date data before continuing."
    )


# ------------------------------------------------------------
# 5. ENSURE LEAD TIME CATEGORY EXISTS
# ------------------------------------------------------------

q1 = df["Delivery_Days"].quantile(0.25)
q3 = df["Delivery_Days"].quantile(0.75)

df["Lead_Time_Category"] = "Moderate"

df.loc[
    df["Delivery_Days"] <= q1,
    "Lead_Time_Category"
] = "Fast"

df.loc[
    df["Delivery_Days"] > q3,
    "Lead_Time_Category"
] = "Slow"


print("\n========== LEAD-TIME THRESHOLDS ==========")

print(f"Fast threshold     : <= {q1:.0f} days")
print(f"Slow threshold     : >  {q3:.0f} days")


# ------------------------------------------------------------
# 6. PRODUCT BASE ANALYSIS
# ------------------------------------------------------------

product_priority = (
    df.groupby("Product Name", observed=True)
    .agg(
        Orders=("Order ID", "nunique"),
        Records=("Order ID", "size"),
        Average_Lead_Time=("Delivery_Days", "mean"),
        Median_Lead_Time=("Delivery_Days", "median"),
        Total_Sales=("Sales", "sum"),
        Total_Gross_Profit=("Gross Profit", "sum")
    )
)


# ------------------------------------------------------------
# 7. SLOW SHIPMENT RATE
# ------------------------------------------------------------

slow_rate = (
    (df["Lead_Time_Category"] == "Slow")
    .astype(int)
    .groupby(df["Product Name"], observed=True)
    .mean()
    * 100
)

product_priority["Slow_Shipment_%"] = slow_rate


# ------------------------------------------------------------
# 8. GROSS PROFIT MARGIN
# ------------------------------------------------------------

product_priority["Gross_Margin_%"] = (
    product_priority["Total_Gross_Profit"]
    / product_priority["Total_Sales"]
    * 100
)


# ------------------------------------------------------------
# 9. DATA-DRIVEN SCORES
# ------------------------------------------------------------

product_priority["Lead_Time_Score"] = (
    product_priority["Average_Lead_Time"]
    .rank(pct=True)
    * 100
)

product_priority["Slow_Rate_Score"] = (
    product_priority["Slow_Shipment_%"]
    .rank(pct=True)
    * 100
)

product_priority["Sales_Score"] = (
    product_priority["Total_Sales"]
    .rank(pct=True)
    * 100
)

product_priority["Profit_Score"] = (
    product_priority["Total_Gross_Profit"]
    .rank(pct=True)
    * 100
)


# ------------------------------------------------------------
# 10. PRIORITY SCORE
# ------------------------------------------------------------

product_priority["Priority_Score"] = (
    product_priority["Lead_Time_Score"] * 0.40
    + product_priority["Slow_Rate_Score"] * 0.30
    + product_priority["Sales_Score"] * 0.20
    + product_priority["Profit_Score"] * 0.10
)


# ------------------------------------------------------------
# 11. PRIORITY CATEGORY
# ------------------------------------------------------------

def assign_priority(score):

    if score >= 75:
        return "High Priority"

    elif score >= 50:
        return "Medium Priority"

    else:
        return "Low Priority"


product_priority["Priority_Category"] = (
    product_priority["Priority_Score"]
    .apply(assign_priority)
)


# ------------------------------------------------------------
# 12. SORT RESULTS
# ------------------------------------------------------------

product_priority = product_priority.sort_values(
    "Priority_Score",
    ascending=False
)


# ------------------------------------------------------------
# 13. DISPLAY RESULTS
# ------------------------------------------------------------

display_columns = [
    "Orders",
    "Records",
    "Average_Lead_Time",
    "Median_Lead_Time",
    "Slow_Shipment_%",
    "Total_Sales",
    "Total_Gross_Profit",
    "Gross_Margin_%",
    "Priority_Score",
    "Priority_Category"
]

print("\n========== PRODUCT PRIORITY TABLE ==========\n")

print(
    product_priority[display_columns]
    .round(2)
    .to_string()
)


# ------------------------------------------------------------
# 14. TOP 5 PRIORITY PRODUCTS
# ------------------------------------------------------------

print("\n========== TOP 5 PRIORITY PRODUCTS ==========\n")

print(
    product_priority[
        [
            "Average_Lead_Time",
            "Slow_Shipment_%",
            "Total_Sales",
            "Total_Gross_Profit",
            "Priority_Score",
            "Priority_Category"
        ]
    ]
    .head(5)
    .round(2)
    .to_string()
)


# ------------------------------------------------------------
# 15. PRIORITY DISTRIBUTION
# ------------------------------------------------------------

print("\n========== PRIORITY DISTRIBUTION ==========\n")

print(
    product_priority["Priority_Category"]
    .value_counts()
    .to_string()
)


# ------------------------------------------------------------
# 16. VALIDATION
# ------------------------------------------------------------

print("\n========== FINAL VALIDATION ==========")

print(
    f"Products analyzed          : {len(product_priority)}"
)

print(
    f"Missing Priority Scores    : "
    f"{product_priority['Priority_Score'].isna().sum()}"
)

print(
    f"Missing Priority Categories: "
    f"{product_priority['Priority_Category'].isna().sum()}"
)

print(
    f"Total dataset records      : {len(df):,}"
)

print("\n✓ STEP 16 COMPLETED SUCCESSFULLY")
````

````python
# ============================================================
# STEP 17: VALIDATED PRODUCT → FACTORY MAPPING
# FINAL CORRECTED VERSION
# ============================================================

import pandas as pd
from pathlib import Path
import re
import unicodedata

print("=" * 70)
print("STEP 17: VALIDATED PRODUCT → FACTORY MAPPING")
print("=" * 70)


# ============================================================
# 1. LOAD DATASET SAFELY
# ============================================================

if "df" not in globals():

    print("\nDataframe 'df' not found. Loading CSV...")

    possible_paths = [
        Path("../data/Nassau Candy Distributor.csv"),
        Path("data/Nassau Candy Distributor.csv"),
        Path("Nassau Candy Distributor.csv")
    ]

    csv_path = next(
        (p for p in possible_paths if p.exists()),
        None
    )

    if csv_path is None:
        raise FileNotFoundError(
            "Nassau Candy Distributor.csv was not found."
        )

    df = pd.read_csv(csv_path)

    print(f"Dataset loaded from: {csv_path}")

else:

    print("\nExisting dataframe 'df' found.")


# ============================================================
# 2. BASIC VALIDATION
# ============================================================

required_columns = [
    "Product Name",
    "Order ID",
    "Sales",
    "Units",
    "Gross Profit"
]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:

    raise KeyError(
        f"Required columns missing from dataset: {missing_columns}"
    )


print(f"Dataset rows       : {len(df):,}")
print(f"Unique products    : {df['Product Name'].nunique()}")


# ============================================================
# 3. CANONICAL PRODUCT NAME FUNCTION
# ============================================================
# This handles formatting differences such as:
#
# Wonka Bar -Scrumdiddlyumptious
# Wonka Bar - Scrumdiddlyumptious
#
# Both become:
# Wonka Bar - Scrumdiddlyumptious
#
# IMPORTANT:
# Original Product Name column is NOT changed.
# Only the temporary matching key is normalized.
# ============================================================

def canonical_product_name(value):

    if pd.isna(value):
        return ""

    text = str(value)

    # Unicode normalization
    text = unicodedata.normalize("NFKC", text)

    # Convert non-breaking spaces to normal spaces
    text = text.replace("\u00A0", " ")

    # Remove leading/trailing whitespace
    text = text.strip()

    # Normalize spaces around hyphens
    text = re.sub(r"\s*-\s*", " - ", text)

    # Remove repeated spaces
    text = re.sub(r"\s+", " ", text)

    return text


# Create temporary matching key
df["_Product_Key"] = df["Product Name"].apply(
    canonical_product_name
)


# ============================================================
# 4. FACTORY MAPPING
# ============================================================

factory_mapping = {

    # --------------------------------------------------------
    # LOT'S O' NUTS
    # --------------------------------------------------------
    "Wonka Bar - Nutty Crunch Surprise": "Lot's O' Nuts",
    "Wonka Bar - Fudge Mallows": "Lot's O' Nuts",
    "Wonka Bar - Scrumdiddlyumptious": "Lot's O' Nuts",

    # --------------------------------------------------------
    # WICKED CHOCCY'S
    # --------------------------------------------------------
    "Wonka Bar - Milk Chocolate": "Wicked Choccy's",
    "Wonka Bar - Triple Dazzle Caramel": "Wicked Choccy's",

    # --------------------------------------------------------
    # SUGAR SHACK
    # --------------------------------------------------------
    "Laffy Taffy": "Sugar Shack",
    "SweeTARTS": "Sugar Shack",
    "Nerds": "Sugar Shack",
    "Fun Dip": "Sugar Shack",
    "Fizzy Lifting Drinks": "Sugar Shack",

    # --------------------------------------------------------
    # SECRET FACTORY
    # --------------------------------------------------------
    "Everlasting Gobstopper": "Secret Factory",
    "Lickable Wallpaper": "Secret Factory",
    "Wonka Gum": "Secret Factory",

    # --------------------------------------------------------
    # THE OTHER FACTORY
    # --------------------------------------------------------
    "Hair Toffee": "The Other Factory",
    "Kazookles": "The Other Factory"
}


# ============================================================
# 5. NORMALIZE MAPPING KEYS
# ============================================================

normalized_mapping = {
    canonical_product_name(product): factory
    for product, factory in factory_mapping.items()
}


# ============================================================
# 6. VALIDATE MAPPING COUNT
# ============================================================

if len(factory_mapping) != 15:

    raise ValueError(
        f"ERROR: Expected 15 mappings, "
        f"but found {len(factory_mapping)}."
    )


if len(normalized_mapping) != 15:

    raise ValueError(
        "ERROR: Duplicate product names detected "
        "after normalization."
    )


print("\nMapping products  :", len(factory_mapping))
print("Mapping factories :", len(set(factory_mapping.values())))


# ============================================================
# 7. COMPARE DATASET PRODUCTS WITH MAPPING
# ============================================================

dataset_products = set(
    df["_Product_Key"].dropna().unique()
)

mapping_products = set(
    normalized_mapping.keys()
)


missing_products = dataset_products - mapping_products
extra_products = mapping_products - dataset_products


# ------------------------------------------------------------
# Missing products
# ------------------------------------------------------------

if missing_products:

    print("\n❌ PRODUCTS WITHOUT FACTORY MAPPING:")

    for product in sorted(missing_products):
        print("   -", product)

    raise ValueError(
        "Factory mapping validation failed because "
        "one or more dataset products are unmapped."
    )


# ------------------------------------------------------------
# Extra mapping products
# ------------------------------------------------------------

if extra_products:

    print("\n❌ EXTRA PRODUCTS IN FACTORY MAPPING:")

    for product in sorted(extra_products):
        print("   -", product)

    raise ValueError(
        "Factory mapping contains products that "
        "do not exist in the dataset."
    )


# ============================================================
# 8. ASSIGN FACTORY
# ============================================================

df["Factory"] = df["_Product_Key"].map(
    normalized_mapping
)


# ============================================================
# 9. CHECK UNMAPPED ROWS
# ============================================================

unmapped_rows = df["Factory"].isna().sum()


if unmapped_rows != 0:

    raise ValueError(
        f"ERROR: {unmapped_rows:,} rows have no factory."
    )


# ============================================================
# 10. REMOVE TEMPORARY COLUMN
# ============================================================

df.drop(
    columns=["_Product_Key"],
    inplace=True
)


# ============================================================
# 11. PRODUCT → FACTORY VALIDATION TABLE
# ============================================================

product_factory_check = (
    df[["Product Name", "Factory"]]
    .drop_duplicates()
    .sort_values(
        ["Factory", "Product Name"]
    )
    .reset_index(drop=True)
)


# ============================================================
# 12. VERIFY EACH PRODUCT HAS EXACTLY ONE FACTORY
# ============================================================

factory_count_per_product = (
    product_factory_check
    .groupby("Product Name")["Factory"]
    .nunique()
)


invalid_products = (
    factory_count_per_product[
        factory_count_per_product != 1
    ]
)


if len(invalid_products) != 0:

    print("\n❌ PRODUCTS WITH MULTIPLE FACTORIES:")

    print(invalid_products)

    raise ValueError(
        "A product is assigned to more than one factory."
    )


# ============================================================
# 13. DISPLAY PRODUCT → FACTORY MAPPING
# ============================================================

print("\n")
print("=" * 70)
print("PRODUCT → FACTORY MAPPING")
print("=" * 70)

print(
    product_factory_check.to_string(index=False)
)


# ============================================================
# 14. FACTORY PERFORMANCE SUMMARY
# ============================================================

factory_summary = (
    df.groupby("Factory")
      .agg(
          Products=("Product Name", "nunique"),
          Records=("Product Name", "size"),
          Orders=("Order ID", "nunique"),
          Sales=("Sales", "sum"),
          Units=("Units", "sum"),
          Gross_Profit=("Gross Profit", "sum")
      )
      .reset_index()
      .sort_values(
          "Sales",
          ascending=False
      )
)


print("\n")
print("=" * 70)
print("FACTORY SUMMARY")
print("=" * 70)

print(
    factory_summary.to_string(index=False)
)


# ============================================================
# 15. FINAL AUTOMATED VALIDATION
# ============================================================

assert df["Factory"].notna().all(), \
    "Validation failed: NULL factory values exist."

assert df["Product Name"].nunique() == 15, \
    "Validation failed: dataset should contain 15 products."

assert len(factory_mapping) == 15, \
    "Validation failed: mapping should contain 15 products."

assert df["Factory"].nunique() == 5, \
    "Validation failed: dataset should contain 5 factories."

assert len(product_factory_check) == 15, \
    "Validation failed: product-factory table should contain 15 products."

assert (factory_count_per_product == 1).all(), \
    "Validation failed: every product must have exactly one factory."


# ============================================================
# 16. SUCCESS
# ============================================================

print("\n")
print("=" * 70)
print("✅ STEP 17 COMPLETED SUCCESSFULLY")
print("=" * 70)

print("\nValidation Results:")
print(f"✔ Dataset rows              : {len(df):,}")
print(f"✔ Dataset products          : {df['Product Name'].nunique()}")
print(f"✔ Mapped products           : {len(factory_mapping)}")
print(f"✔ Factories                 : {df['Factory'].nunique()}")
print(f"✔ Unmapped rows             : {df['Factory'].isna().sum():,}")
print(f"✔ Products with >1 factory  : {len(invalid_products)}")

print("\n🎯 Factory mapping is VALIDATED.")
````

````python
# ============================================================
# STEP 18: FACTORY PERFORMANCE BASELINE ANALYSIS
# ============================================================

import pandas as pd
from pathlib import Path

print("=" * 70)
print("STEP 18: FACTORY PERFORMANCE BASELINE ANALYSIS")
print("=" * 70)


# ============================================================
# 1. ENSURE DATAFRAME EXISTS
# ============================================================

if "df" not in globals():

    print("\nDataframe 'df' not found. Loading dataset...")

    possible_paths = [
        Path("../data/Nassau Candy Distributor.csv"),
        Path("data/Nassau Candy Distributor.csv"),
        Path("Nassau Candy Distributor.csv")
    ]

    csv_path = next(
        (p for p in possible_paths if p.exists()),
        None
    )

    if csv_path is None:
        raise FileNotFoundError(
            "Nassau Candy Distributor.csv was not found."
        )

    df = pd.read_csv(csv_path)

    print(f"Dataset loaded from: {csv_path}")


else:

    print("\nExisting dataframe 'df' found.")


# ============================================================
# 2. REQUIRED COLUMNS CHECK
# ============================================================

required_columns = [
    "Factory",
    "Product Name",
    "Order ID",
    "Sales",
    "Units",
    "Gross Profit"
]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:

    raise KeyError(
        f"Required columns are missing: {missing_columns}"
    )


# ============================================================
# 3. ENSURE DELIVERY_DAYS EXISTS
# ============================================================

if "Delivery_Days" not in df.columns:

    if "Order Date" not in df.columns or "Ship Date" not in df.columns:

        raise KeyError(
            "Delivery_Days is missing and Order Date/Ship Date "
            "are not available."
        )

    # Handle both datetime and raw string columns safely
    if not pd.api.types.is_datetime64_any_dtype(df["Order Date"]):

        df["Order Date"] = pd.to_datetime(
            df["Order Date"],
            format="%d-%m-%Y",
            errors="raise"
        )

    if not pd.api.types.is_datetime64_any_dtype(df["Ship Date"]):

        df["Ship Date"] = pd.to_datetime(
            df["Ship Date"],
            format="%d-%m-%Y",
            errors="raise"
        )

    df["Delivery_Days"] = (
        df["Ship Date"] - df["Order Date"]
    ).dt.days


# ============================================================
# 4. VALIDATE DELIVERY DAYS
# ============================================================

if df["Delivery_Days"].isna().any():

    raise ValueError(
        "Delivery_Days contains missing values."
    )

if (df["Delivery_Days"] < 0).any():

    raise ValueError(
        "Negative Delivery_Days detected."
    )


# ============================================================
# 5. VALIDATE FACTORY ASSIGNMENT
# ============================================================

if df["Factory"].isna().any():

    raise ValueError(
        "Factory column contains unmapped rows."
    )

factory_count = df["Factory"].nunique()

if factory_count != 5:

    raise ValueError(
        f"Expected 5 factories, but found {factory_count}."
    )


print(f"\nRows                  : {len(df):,}")
print(f"Factories             : {df['Factory'].nunique()}")
print(f"Unmapped factory rows : {df['Factory'].isna().sum():,}")


# ============================================================
# 6. FACTORY PERFORMANCE SUMMARY
# ============================================================

factory_performance = (
    df.groupby("Factory")
      .agg(
          Products=("Product Name", "nunique"),
          Records=("Product Name", "size"),
          Orders=("Order ID", "nunique"),
          Sales=("Sales", "sum"),
          Units=("Units", "sum"),
          Gross_Profit=("Gross Profit", "sum"),
          Avg_Lead_Time=("Delivery_Days", "mean"),
          Median_Lead_Time=("Delivery_Days", "median")
      )
      .reset_index()
)


# ============================================================
# 7. CALCULATE PROFIT MARGIN
# ============================================================

factory_performance["Gross_Profit_Margin_%"] = (
    factory_performance["Gross_Profit"]
    / factory_performance["Sales"]
    * 100
)


# ============================================================
# 8. CALCULATE SLOW-SHIPMENT RATE
# ============================================================
# Slow threshold is based on the previously established
# dataset-level Q3 lead-time threshold.

q3_lead_time = df["Delivery_Days"].quantile(0.75)

factory_performance["Slow_Shipment_%"] = (
    df.assign(
        Slow_Shipment=df["Delivery_Days"] > q3_lead_time
    )
    .groupby("Factory")["Slow_Shipment"]
    .mean()
    .reindex(factory_performance["Factory"])
    .values
    * 100
)


# ============================================================
# 9. ROUND NUMERIC VALUES FOR DISPLAY
# ============================================================

display_table = factory_performance.copy()

display_table["Sales"] = display_table["Sales"].round(2)
display_table["Gross_Profit"] = display_table["Gross_Profit"].round(2)
display_table["Avg_Lead_Time"] = display_table["Avg_Lead_Time"].round(2)
display_table["Median_Lead_Time"] = display_table["Median_Lead_Time"].round(2)
display_table["Gross_Profit_Margin_%"] = (
    display_table["Gross_Profit_Margin_%"].round(2)
)
display_table["Slow_Shipment_%"] = (
    display_table["Slow_Shipment_%"].round(2)
)


# ============================================================
# 10. RANK FACTORIES
# ============================================================

display_table["Sales_Rank"] = (
    display_table["Sales"]
    .rank(method="min", ascending=False)
    .astype(int)
)

display_table["Profit_Rank"] = (
    display_table["Gross_Profit"]
    .rank(method="min", ascending=False)
    .astype(int)
)

display_table["Lead_Time_Rank"] = (
    display_table["Avg_Lead_Time"]
    .rank(method="min", ascending=True)
    .astype(int)
)


# ============================================================
# 11. SORT BY SALES
# ============================================================

display_table = (
    display_table
    .sort_values("Sales", ascending=False)
    .reset_index(drop=True)
)


# ============================================================
# 12. DISPLAY FACTORY PERFORMANCE
# ============================================================

print("\n")
print("=" * 70)
print("FACTORY PERFORMANCE SUMMARY")
print("=" * 70)

print(
    display_table.to_string(index=False)
)


# ============================================================
# 13. IDENTIFY BEST / WORST FACTORIES
# ============================================================

best_sales_factory = (
    factory_performance.loc[
        factory_performance["Sales"].idxmax(),
        "Factory"
    ]
)

best_profit_factory = (
    factory_performance.loc[
        factory_performance["Gross_Profit"].idxmax(),
        "Factory"
    ]
)

fastest_factory = (
    factory_performance.loc[
        factory_performance["Avg_Lead_Time"].idxmin(),
        "Factory"
    ]
)

slowest_factory = (
    factory_performance.loc[
        factory_performance["Avg_Lead_Time"].idxmax(),
        "Factory"
    ]
)

highest_margin_factory = (
    factory_performance.loc[
        factory_performance["Gross_Profit_Margin_%"].idxmax(),
        "Factory"
    ]
)


# ============================================================
# 14. KEY FINDINGS
# ============================================================

print("\n")
print("=" * 70)
print("KEY FACTORY FINDINGS")
print("=" * 70)

print(f"Highest Sales Factory       : {best_sales_factory}")
print(f"Highest Gross Profit Factory: {best_profit_factory}")
print(f"Fastest Factory             : {fastest_factory}")
print(f"Slowest Factory             : {slowest_factory}")
print(f"Highest Margin Factory      : {highest_margin_factory}")


# ============================================================
# 15. AUTOMATED VALIDATION
# ============================================================

assert len(factory_performance) == 5

assert factory_performance["Sales"].notna().all()

assert factory_performance["Gross_Profit"].notna().all()

assert factory_performance["Gross_Profit_Margin_%"].notna().all()

assert factory_performance["Avg_Lead_Time"].notna().all()

assert factory_performance["Slow_Shipment_%"].notna().all()

assert (factory_performance["Gross_Profit_Margin_%"] >= 0).all()

assert (factory_performance["Gross_Profit_Margin_%"] <= 100).all()

assert (factory_performance["Slow_Shipment_%"] >= 0).all()

assert (factory_performance["Slow_Shipment_%"] <= 100).all()


# ============================================================
# 16. SUCCESS
# ============================================================

print("\n")
print("=" * 70)
print("✅ STEP 18 COMPLETED SUCCESSFULLY")
print("=" * 70)

print(
    "\nFactory baseline performance has been calculated "
    "and validated for all 5 factories."
)
````

````python
# ============================================================
# STEP 19: PRODUCT × FACTORY OPERATIONAL ANALYSIS
# ============================================================

import pandas as pd
from pathlib import Path

print("=" * 75)
print("STEP 19: PRODUCT × FACTORY OPERATIONAL ANALYSIS")
print("=" * 75)


# ============================================================
# 1. ENSURE DATAFRAME EXISTS
# ============================================================

if "df" not in globals():

    print("\nDataframe 'df' not found. Loading dataset...")

    possible_paths = [
        Path("../data/Nassau Candy Distributor.csv"),
        Path("data/Nassau Candy Distributor.csv"),
        Path("Nassau Candy Distributor.csv")
    ]

    csv_path = next(
        (p for p in possible_paths if p.exists()),
        None
    )

    if csv_path is None:
        raise FileNotFoundError(
            "Nassau Candy Distributor.csv was not found."
        )

    df = pd.read_csv(csv_path)

    # Correct raw CSV date format
    if "Order Date" in df.columns:
        df["Order Date"] = pd.to_datetime(
            df["Order Date"],
            format="%d-%m-%Y",
            errors="raise"
        )

    if "Ship Date" in df.columns:
        df["Ship Date"] = pd.to_datetime(
            df["Ship Date"],
            format="%d-%m-%Y",
            errors="raise"
        )

    if "Delivery_Days" not in df.columns:
        df["Delivery_Days"] = (
            df["Ship Date"] - df["Order Date"]
        ).dt.days

    print(f"Dataset loaded from: {csv_path}")

else:

    print("\nExisting dataframe 'df' found.")


# ============================================================
# 2. REQUIRED COLUMNS
# ============================================================

required_columns = [
    "Product Name",
    "Factory",
    "Order ID",
    "Sales",
    "Units",
    "Gross Profit",
    "Delivery_Days"
]

missing_columns = [
    column
    for column in required_columns
    if column not in df.columns
]

if missing_columns:

    raise KeyError(
        f"Required columns missing: {missing_columns}"
    )


# ============================================================
# 3. DATA VALIDATION
# ============================================================

if df["Factory"].isna().any():

    raise ValueError(
        "Factory contains unmapped rows."
    )


if df["Product Name"].isna().any():

    raise ValueError(
        "Product Name contains missing values."
    )


if df["Delivery_Days"].isna().any():

    raise ValueError(
        "Delivery_Days contains missing values."
    )


if (df["Delivery_Days"] < 0).any():

    raise ValueError(
        "Negative Delivery_Days detected."
    )


# ============================================================
# 4. DATASET-LEVEL LEAD TIME THRESHOLD
# ============================================================

q3_lead_time = df["Delivery_Days"].quantile(0.75)

print(f"\nDataset Q3 Lead Time : {q3_lead_time:.2f} days")


# ============================================================
# 5. CREATE SLOW SHIPMENT FLAG
# ============================================================

analysis_df = df.copy()

analysis_df["_Slow_Shipment"] = (
    analysis_df["Delivery_Days"] > q3_lead_time
)


# ============================================================
# 6. PRODUCT × FACTORY SUMMARY
# ============================================================

product_factory_analysis = (
    analysis_df
    .groupby(
        ["Product Name", "Factory"],
        as_index=False
    )
    .agg(
        Orders=("Order ID", "nunique"),
        Records=("Product Name", "size"),
        Sales=("Sales", "sum"),
        Units=("Units", "sum"),
        Gross_Profit=("Gross Profit", "sum"),
        Avg_Lead_Time=("Delivery_Days", "mean"),
        Median_Lead_Time=("Delivery_Days", "median"),
        Slow_Shipment_Rate=("_Slow_Shipment", "mean")
    )
)


# ============================================================
# 7. CALCULATE GROSS PROFIT MARGIN
# ============================================================

product_factory_analysis["Gross_Profit_Margin_%"] = (
    product_factory_analysis["Gross_Profit"]
    / product_factory_analysis["Sales"]
    * 100
)


# Convert slow shipment rate to percentage
product_factory_analysis["Slow_Shipment_%"] = (
    product_factory_analysis["Slow_Shipment_Rate"] * 100
)


product_factory_analysis.drop(
    columns=["Slow_Shipment_Rate"],
    inplace=True
)


# ============================================================
# 8. ROUND VALUES
# ============================================================

numeric_columns = [
    "Sales",
    "Gross_Profit",
    "Avg_Lead_Time",
    "Median_Lead_Time",
    "Gross_Profit_Margin_%",
    "Slow_Shipment_%"
]

for column in numeric_columns:

    product_factory_analysis[column] = (
        product_factory_analysis[column].round(2)
    )


# ============================================================
# 9. OPERATIONAL PRIORITY SCORE
# ============================================================
# This is NOT an ML prediction.
# It is only a baseline screening indicator.
#
# Higher priority is assigned to products with:
# - longer lead time
# - higher slow-shipment rate
# - meaningful sales volume
#
# The score is intentionally transparent and will NOT be used
# as the final reallocation recommendation.
# ============================================================

lead_min = product_factory_analysis["Avg_Lead_Time"].min()
lead_max = product_factory_analysis["Avg_Lead_Time"].max()

slow_min = product_factory_analysis["Slow_Shipment_%"].min()
slow_max = product_factory_analysis["Slow_Shipment_%"].max()

sales_min = product_factory_analysis["Sales"].min()
sales_max = product_factory_analysis["Sales"].max()


def min_max_scale(series, minimum, maximum):

    if maximum == minimum:
        return pd.Series(
            0.0,
            index=series.index
        )

    return (
        (series - minimum)
        / (maximum - minimum)
    )


lead_score = min_max_scale(
    product_factory_analysis["Avg_Lead_Time"],
    lead_min,
    lead_max
)

slow_score = min_max_scale(
    product_factory_analysis["Slow_Shipment_%"],
    slow_min,
    slow_max
)

sales_score = min_max_scale(
    product_factory_analysis["Sales"],
    sales_min,
    sales_max
)


product_factory_analysis["Operational_Priority_Score"] = (
    0.45 * lead_score
    + 0.35 * slow_score
    + 0.20 * sales_score
)


# ============================================================
# 10. PRIORITY CATEGORY
# ============================================================

product_factory_analysis["Operational_Priority"] = pd.cut(
    product_factory_analysis["Operational_Priority_Score"],
    bins=[
        -float("inf"),
        0.33,
        0.66,
        float("inf")
    ],
    labels=[
        "Low",
        "Medium",
        "High"
    ],
    include_lowest=True
)


product_factory_analysis[
    "Operational_Priority_Score"
] = (
    product_factory_analysis[
        "Operational_Priority_Score"
    ].round(3)
)


# ============================================================
# 11. SORT BY PRIORITY
# ============================================================

product_factory_analysis = (
    product_factory_analysis
    .sort_values(
        [
            "Operational_Priority_Score",
            "Sales"
        ],
        ascending=[False, False]
    )
    .reset_index(drop=True)
)


# ============================================================
# 12. DISPLAY COMPLETE ANALYSIS
# ============================================================

print("\n")
print("=" * 75)
print("PRODUCT × FACTORY OPERATIONAL ANALYSIS")
print("=" * 75)

print(
    product_factory_analysis.to_string(index=False)
)


# ============================================================
# 13. TOP OPERATIONAL PRIORITY PRODUCTS
# ============================================================

top_priority = (
    product_factory_analysis
    .head(10)
    .copy()
)


print("\n")
print("=" * 75)
print("TOP 10 OPERATIONAL PRIORITY PRODUCTS")
print("=" * 75)

print(
    top_priority[
        [
            "Product Name",
            "Factory",
            "Sales",
            "Avg_Lead_Time",
            "Slow_Shipment_%",
            "Gross_Profit_Margin_%",
            "Operational_Priority_Score",
            "Operational_Priority"
        ]
    ].to_string(index=False)
)


# ============================================================
# 14. FACTORY × PRODUCT COUNT
# ============================================================

factory_product_count = (
    product_factory_analysis
    .groupby("Factory")
    .agg(
        Products=("Product Name", "nunique"),
        Sales=("Sales", "sum"),
        Gross_Profit=("Gross_Profit", "sum"),
        Avg_Lead_Time=("Avg_Lead_Time", "mean")
    )
    .reset_index()
)


factory_product_count["Gross_Profit_Margin_%"] = (
    factory_product_count["Gross_Profit"]
    / factory_product_count["Sales"]
    * 100
)


factory_product_count["Avg_Lead_Time"] = (
    factory_product_count["Avg_Lead_Time"].round(2)
)

factory_product_count["Gross_Profit_Margin_%"] = (
    factory_product_count["Gross_Profit_Margin_%"].round(2)
)


print("\n")
print("=" * 75)
print("FACTORY PRODUCT PORTFOLIO")
print("=" * 75)

print(
    factory_product_count.to_string(index=False)
)


# ============================================================
# 15. AUTOMATED VALIDATION
# ============================================================

# Exactly 15 product-factory combinations should exist
assert len(product_factory_analysis) == 15, (
    "Validation failed: expected exactly 15 "
    "product-factory combinations."
)


# Every product should have exactly one current factory
product_factory_counts = (
    product_factory_analysis
    .groupby("Product Name")["Factory"]
    .nunique()
)

assert (
    product_factory_counts == 1
).all(), (
    "Validation failed: a product has "
    "more than one current factory."
)


# Exactly 5 factories
assert (
    product_factory_analysis["Factory"].nunique() == 5
), (
    "Validation failed: expected 5 factories."
)


# No missing values in important metrics
assert (
    product_factory_analysis[
        [
            "Sales",
            "Units",
            "Gross_Profit",
            "Avg_Lead_Time",
            "Gross_Profit_Margin_%",
            "Slow_Shipment_%",
            "Operational_Priority_Score"
        ]
    ]
    .notna()
    .all()
    .all()
), (
    "Validation failed: missing analytical values detected."
)


# Margin must be between 0 and 100
assert (
    product_factory_analysis[
        "Gross_Profit_Margin_%"
    ].between(0, 100).all()
), (
    "Validation failed: invalid gross profit margin."
)


# Slow shipment percentage must be between 0 and 100
assert (
    product_factory_analysis[
        "Slow_Shipment_%"
    ].between(0, 100).all()
), (
    "Validation failed: invalid slow shipment percentage."
)


# Priority score must be between 0 and 1
assert (
    product_factory_analysis[
        "Operational_Priority_Score"
    ].between(0, 1).all()
), (
    "Validation failed: operational priority score "
    "outside expected range."
)


# ============================================================
# 16. SUCCESS
# ============================================================

print("\n")
print("=" * 75)
print("✅ STEP 19 COMPLETED SUCCESSFULLY")
print("=" * 75)

print(
    "\nProduct × Factory operational baseline "
    "has been calculated and validated."
)

print(
    "\nImportant: Operational Priority is a screening metric "
    "only. It is NOT the final factory-reallocation decision."
)
````

````python
# ============================================================
# STEP 22: FACTORY × CUSTOMER REGION OPERATIONAL ANALYSIS
# ============================================================

import pandas as pd
import numpy as np
import os
import re
import unicodedata

print("=" * 80)
print("STEP 22: FACTORY × CUSTOMER REGION OPERATIONAL ANALYSIS")
print("=" * 80)


# ------------------------------------------------------------
# 1. LOAD DATA SAFELY
# ------------------------------------------------------------

data_path = "../data/Nassau Candy Distributor.csv"

if not os.path.exists(data_path):
    raise FileNotFoundError(
        f"ERROR: Dataset not found at: {data_path}"
    )

df_step22 = pd.read_csv(data_path)

print("\n✓ Raw dataset loaded successfully.")
print(f"✓ Records loaded: {len(df_step22):,}")


# ------------------------------------------------------------
# 2. REQUIRED COLUMN VALIDATION
# ------------------------------------------------------------

required_columns = [
    "Order ID",
    "Order Date",
    "Ship Date",
    "Customer ID",
    "Region",
    "Product Name",
    "Sales",
    "Units",
    "Gross Profit",
    "Cost"
]

missing_columns = [
    col for col in required_columns
    if col not in df_step22.columns
]

print("\nRequired columns:")
print(required_columns)

print("\nMissing columns:")
print(missing_columns)

assert not missing_columns, (
    f"ERROR: Required columns are missing: {missing_columns}"
)

print("✓ Required columns validated.")


# ------------------------------------------------------------
# 3. PARSE DATES FROM RAW CSV
# ------------------------------------------------------------

df_step22["Order Date"] = pd.to_datetime(
    df_step22["Order Date"],
    format="%d-%m-%Y",
    errors="raise"
)

df_step22["Ship Date"] = pd.to_datetime(
    df_step22["Ship Date"],
    format="%d-%m-%Y",
    errors="raise"
)

df_step22["Delivery_Days"] = (
    df_step22["Ship Date"] - df_step22["Order Date"]
).dt.days

assert df_step22["Order Date"].notna().all()
assert df_step22["Ship Date"].notna().all()
assert df_step22["Delivery_Days"].notna().all()
assert (df_step22["Delivery_Days"] >= 0).all()

print("✓ Dates parsed successfully.")
print("✓ Delivery_Days calculated successfully.")
print("✓ No negative delivery days found.")


# ------------------------------------------------------------
# 4. PRODUCT NAME NORMALIZATION
# ------------------------------------------------------------

def canonical_product_name(value):
    if pd.isna(value):
        return ""

    text = str(value)

    text = unicodedata.normalize("NFKC", text)
    text = text.replace("\u00A0", " ")
    text = text.strip()

    # Normalize spaces around hyphen
    text = re.sub(r"\s*-\s*", " - ", text)

    # Normalize repeated spaces
    text = re.sub(r"\s+", " ", text)

    return text


df_step22["_Product_Key"] = (
    df_step22["Product Name"]
    .apply(canonical_product_name)
)


# ------------------------------------------------------------
# 5. VALIDATED PRODUCT → FACTORY MAPPING
# ------------------------------------------------------------

product_factory_mapping = {

    # Lot's O' Nuts
    "Wonka Bar - Nutty Crunch Surprise": "Lot's O' Nuts",
    "Wonka Bar - Fudge Mallows": "Lot's O' Nuts",
    "Wonka Bar - Scrumdiddlyumptious": "Lot's O' Nuts",

    # Wicked Choccy's
    "Wonka Bar - Milk Chocolate": "Wicked Choccy's",
    "Wonka Bar - Triple Dazzle Caramel": "Wicked Choccy's",

    # Sugar Shack
    "Laffy Taffy": "Sugar Shack",
    "SweeTARTS": "Sugar Shack",
    "Nerds": "Sugar Shack",
    "Fun Dip": "Sugar Shack",
    "Fizzy Lifting Drinks": "Sugar Shack",

    # Secret Factory
    "Everlasting Gobstopper": "Secret Factory",
    "Lickable Wallpaper": "Secret Factory",
    "Wonka Gum": "Secret Factory",

    # The Other Factory
    "Hair Toffee": "The Other Factory",
    "Kazookles": "The Other Factory"
}


# Normalize mapping keys too
normalized_mapping = {
    canonical_product_name(product): factory
    for product, factory in product_factory_mapping.items()
}


df_step22["Factory"] = (
    df_step22["_Product_Key"]
    .map(normalized_mapping)
)


# ------------------------------------------------------------
# 6. FACTORY ASSIGNMENT VALIDATION
# ------------------------------------------------------------

missing_factory_rows = df_step22["Factory"].isna().sum()

print("\nFactory assignment validation:")
print(f"Unique products : {df_step22['Product Name'].nunique()}")
print(f"Unique factories: {df_step22['Factory'].nunique()}")
print(f"Missing factory rows: {missing_factory_rows}")

assert missing_factory_rows == 0, (
    "ERROR: Some products could not be mapped to a factory."
)

expected_factories = {
    "Lot's O' Nuts",
    "Wicked Choccy's",
    "Sugar Shack",
    "Secret Factory",
    "The Other Factory"
}

actual_factories = set(df_step22["Factory"].unique())

assert actual_factories == expected_factories, (
    f"ERROR: Factory set mismatch.\n"
    f"Expected: {expected_factories}\n"
    f"Actual: {actual_factories}"
)

assert df_step22["Product Name"].nunique() == 15
assert df_step22["Factory"].nunique() == 5

print("✓ All 15 products mapped.")
print("✓ Exactly 5 validated factories detected.")


# ------------------------------------------------------------
# 7. DATA-DRIVEN LEAD-TIME THRESHOLD
# ------------------------------------------------------------

lead_time_q3 = df_step22["Delivery_Days"].quantile(0.75)

df_step22["Slow_Shipment"] = (
    df_step22["Delivery_Days"] > lead_time_q3
)

print("\nLead-time threshold:")
print(f"Q3 lead time: {lead_time_q3:.0f} days")
print("✓ Slow shipment flag created.")


# ------------------------------------------------------------
# 8. FACTORY × REGION SUMMARY
# ------------------------------------------------------------

factory_region_summary = (
    df_step22
    .groupby(["Factory", "Region"])
    .agg(
        Customers=("Customer ID", "nunique"),
        Orders=("Order ID", "nunique"),
        Records=("Order ID", "size"),
        Sales=("Sales", "sum"),
        Units=("Units", "sum"),
        Gross_Profit=("Gross Profit", "sum"),
        Cost=("Cost", "sum"),
        Avg_Lead_Days=("Delivery_Days", "mean"),
        Median_Lead_Days=("Delivery_Days", "median"),
        Slow_Shipments=("Slow_Shipment", "sum")
    )
    .reset_index()
)


# ------------------------------------------------------------
# 9. CALCULATE BUSINESS KPIs
# ------------------------------------------------------------

factory_region_summary["Profit_Margin_%"] = np.where(
    factory_region_summary["Sales"] != 0,
    (
        factory_region_summary["Gross_Profit"]
        / factory_region_summary["Sales"]
        * 100
    ),
    0
)

factory_region_summary["Slow_Shipment_%"] = np.where(
    factory_region_summary["Records"] != 0,
    (
        factory_region_summary["Slow_Shipments"]
        / factory_region_summary["Records"]
        * 100
    ),
    0
)

factory_region_summary["Sales_Share_%"] = (
    factory_region_summary["Sales"]
    / df_step22["Sales"].sum()
    * 100
)


# ------------------------------------------------------------
# 10. SORT BY SALES
# ------------------------------------------------------------

factory_region_summary = (
    factory_region_summary
    .sort_values(
        ["Sales", "Orders"],
        ascending=[False, False]
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 11. DISPLAY FACTORY × REGION ANALYSIS
# ------------------------------------------------------------

display_columns = [
    "Factory",
    "Region",
    "Customers",
    "Orders",
    "Records",
    "Sales",
    "Units",
    "Gross_Profit",
    "Cost",
    "Profit_Margin_%",
    "Avg_Lead_Days",
    "Median_Lead_Days",
    "Slow_Shipment_%",
    "Sales_Share_%"
]

print("\n" + "=" * 80)
print("FACTORY × CUSTOMER REGION PERFORMANCE")
print("=" * 80)

print(
    factory_region_summary[display_columns].to_string(
        index=False,
        formatters={
            "Sales": "{:,.2f}".format,
            "Gross_Profit": "{:,.2f}".format,
            "Cost": "{:,.2f}".format,
            "Profit_Margin_%": "{:.2f}".format,
            "Avg_Lead_Days": "{:.2f}".format,
            "Median_Lead_Days": "{:.0f}".format,
            "Slow_Shipment_%": "{:.2f}".format,
            "Sales_Share_%": "{:.2f}".format
        }
    )
)


# ------------------------------------------------------------
# 12. WORST FACTORY-REGION LANES
# ------------------------------------------------------------

worst_lanes = (
    factory_region_summary
    .sort_values(
        ["Avg_Lead_Days", "Slow_Shipment_%"],
        ascending=[False, False]
    )
    .head(10)
    .reset_index(drop=True)
)

print("\n" + "=" * 80)
print("TOP 10 FACTORY × REGION LANES BY LEAD-TIME RISK")
print("=" * 80)

print(
    worst_lanes[
        [
            "Factory",
            "Region",
            "Orders",
            "Sales",
            "Avg_Lead_Days",
            "Median_Lead_Days",
            "Slow_Shipment_%",
            "Profit_Margin_%"
        ]
    ].to_string(
        index=False,
        formatters={
            "Sales": "{:,.2f}".format,
            "Avg_Lead_Days": "{:.2f}".format,
            "Median_Lead_Days": "{:.0f}".format,
            "Slow_Shipment_%": "{:.2f}".format,
            "Profit_Margin_%": "{:.2f}".format
        }
    )
)


# ------------------------------------------------------------
# 13. FACTORY PERFORMANCE ACROSS REGIONS
# ------------------------------------------------------------

factory_summary = (
    df_step22
    .groupby("Factory")
    .agg(
        Regions=("Region", "nunique"),
        Orders=("Order ID", "nunique"),
        Records=("Order ID", "size"),
        Sales=("Sales", "sum"),
        Units=("Units", "sum"),
        Gross_Profit=("Gross Profit", "sum"),
        Avg_Lead_Days=("Delivery_Days", "mean")
    )
    .reset_index()
)

factory_summary["Profit_Margin_%"] = (
    factory_summary["Gross_Profit"]
    / factory_summary["Sales"]
    * 100
)

factory_summary = factory_summary.sort_values(
    "Sales",
    ascending=False
).reset_index(drop=True)

print("\n" + "=" * 80)
print("FACTORY-LEVEL SUMMARY")
print("=" * 80)

print(
    factory_summary.to_string(
        index=False,
        formatters={
            "Sales": "{:,.2f}".format,
            "Gross_Profit": "{:,.2f}".format,
            "Avg_Lead_Days": "{:.2f}".format,
            "Profit_Margin_%": "{:.2f}".format
        }
    )
)


# ------------------------------------------------------------
# 14. RECONCILIATION CHECKS
# ------------------------------------------------------------

factory_region_sales = factory_region_summary["Sales"].sum()
dataset_sales = df_step22["Sales"].sum()

factory_region_orders = factory_region_summary["Orders"].sum()

print("\n" + "=" * 80)
print("STEP 22 VALIDATION")
print("=" * 80)

print(f"Dataset records              : {len(df_step22):,}")
print(f"Factory-region combinations  : {len(factory_region_summary):,}")
print(f"Dataset sales                : {dataset_sales:,.2f}")
print(f"Factory-region sales         : {factory_region_sales:,.2f}")
print(f"Unique dataset orders        : {df_step22['Order ID'].nunique():,}")
print(f"Factory-region order count   : {factory_region_orders:,}")


# Sales must reconcile exactly
assert abs(factory_region_sales - dataset_sales) < 0.01, (
    "ERROR: Factory-region sales do not reconcile with dataset sales."
)

# Factory count
assert df_step22["Factory"].nunique() == 5, (
    "ERROR: Expected exactly 5 factories."
)

# Region count
assert df_step22["Region"].nunique() == 4, (
    "ERROR: Expected exactly 4 regions."
)

# Product count
assert df_step22["Product Name"].nunique() == 15, (
    "ERROR: Expected exactly 15 products."
)

# No missing important values
assert factory_region_summary[
    ["Factory", "Region", "Sales", "Orders", "Avg_Lead_Days"]
].notna().all().all(), (
    "ERROR: Important KPI values contain missing values."
)

# Margin validity
assert factory_region_summary[
    "Profit_Margin_%"
].between(0, 100).all(), (
    "ERROR: Profit margin contains invalid values."
)

# Lead time validity
assert (
    factory_region_summary["Avg_Lead_Days"] >= 0
).all(), (
    "ERROR: Average lead time contains negative values."
)


# ------------------------------------------------------------
# 15. FINAL STATUS
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("STEP 22 STATUS")
print("=" * 80)

print("✓ Factory × Region analysis completed.")
print("✓ Factory assignments validated.")
print("✓ Customer regions validated.")
print("✓ Orders, sales and profit metrics calculated.")
print("✓ Average and median lead time calculated.")
print("✓ Slow-shipment rate calculated.")
print("✓ Factory-region sales reconcile with source dataset.")
print("✓ 5 factories validated.")
print("✓ 4 customer regions validated.")
print("✓ 15 products validated.")
print("✓ No factory coordinates were invented.")
print("✓ Distance calculation remains pending verified coordinates.")

print("\nSTEP 22 completed successfully.")
print("=" * 80)
````

````python
# ============================================================
# STEP 25A — ORDER-LEVEL TRAIN/TEST SPLIT
# FINAL VERSION
# ============================================================

import sys
import subprocess
import importlib.util

# ------------------------------------------------------------
# 0. AUTOMATICALLY CHECK / INSTALL SCIKIT-LEARN
# ------------------------------------------------------------

if importlib.util.find_spec("sklearn") is None:

    print("scikit-learn is not installed.")
    print("Installing scikit-learn in the CURRENT Python environment...")
    
    subprocess.check_call([
        sys.executable,
        "-m",
        "pip",
        "install",
        "scikit-learn"
    ])

    print("✓ scikit-learn installation completed.")


# Import AFTER installation
from sklearn.model_selection import GroupShuffleSplit

import os
import re
import unicodedata
import pandas as pd
import numpy as np


print("=" * 70)
print("STEP 25A — ORDER-LEVEL TRAIN/TEST SPLIT")
print("=" * 70)


# ------------------------------------------------------------
# 1. FIND DATASET
# ------------------------------------------------------------

possible_paths = [
    "../data/Nassau Candy Distributor.csv",
    "/mnt/data/Nassau Candy Distributor.csv",
    "/mnt/data/files/Nassau Candy Distributor.csv"
]

DATA_PATH = None

for path in possible_paths:
    if os.path.exists(path):
        DATA_PATH = path
        break

if DATA_PATH is None:
    raise FileNotFoundError(
        "Nassau Candy Distributor.csv not found."
    )

print("✓ Dataset found:")
print(DATA_PATH)


# ------------------------------------------------------------
# 2. LOAD DATA
# ------------------------------------------------------------

df_ml = pd.read_csv(DATA_PATH)

print(f"✓ Dataset loaded")
print(f"✓ Records: {len(df_ml):,}")


# ------------------------------------------------------------
# 3. REQUIRED COLUMNS
# ------------------------------------------------------------

required_columns = [
    "Order ID",
    "Order Date",
    "Ship Date",
    "Ship Mode",
    "Customer ID",
    "Country/Region",
    "State/Province",
    "Region",
    "Product Name",
    "Division",
    "Units",
    "Sales",
    "Cost",
    "Gross Profit"
]

missing_columns = [
    column
    for column in required_columns
    if column not in df_ml.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}"
    )

print("✓ Required columns verified")
print(f"✓ Missing columns: {missing_columns}")


# ------------------------------------------------------------
# 4. PRODUCT NORMALIZATION FUNCTION
# ------------------------------------------------------------

def canonical_product_name(value):

    if pd.isna(value):
        return ""

    text = unicodedata.normalize(
        "NFKC",
        str(value)
    )

    text = text.replace("\u00A0", " ")
    text = text.strip()

    text = re.sub(
        r"\s*-\s*",
        " - ",
        text
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    )

    return text


# ------------------------------------------------------------
# 5. VALIDATED PRODUCT → FACTORY MAPPING
# ------------------------------------------------------------

product_factory_mapping = {

    "Wonka Bar - Nutty Crunch Surprise":
        "Lot’s O’ Nuts",

    "Wonka Bar - Fudge Mallows":
        "Lot’s O’ Nuts",

    "Wonka Bar - Scrumdiddlyumptious":
        "Lot’s O’ Nuts",

    "Wonka Bar - Milk Chocolate":
        "Wicked Choccy’s",

    "Wonka Bar - Triple Dazzle Caramel":
        "Wicked Choccy’s",

    "Laffy Taffy":
        "Sugar Shack",

    "SweeTARTS":
        "Sugar Shack",

    "Nerds":
        "Sugar Shack",

    "Fun Dip":
        "Sugar Shack",

    "Fizzy Lifting Drinks":
        "Sugar Shack",

    "Everlasting Gobstopper":
        "Secret Factory",

    "Lickable Wallpaper":
        "Secret Factory",

    "Wonka Gum":
        "Secret Factory",

    "Hair Toffee":
        "The Other Factory",

    "Kazookles":
        "The Other Factory"
}


normalized_factory_mapping = {
    canonical_product_name(product): factory
    for product, factory
    in product_factory_mapping.items()
}


# ------------------------------------------------------------
# 6. ASSIGN FACTORY
# ------------------------------------------------------------

df_ml["_Product_Key"] = (
    df_ml["Product Name"]
    .map(canonical_product_name)
)

df_ml["Factory"] = (
    df_ml["_Product_Key"]
    .map(normalized_factory_mapping)
)


# ------------------------------------------------------------
# 7. FACTORY VALIDATION
# ------------------------------------------------------------

unmapped_rows = df_ml["Factory"].isna().sum()

if unmapped_rows > 0:

    unmapped_products = (
        df_ml.loc[
            df_ml["Factory"].isna(),
            "Product Name"
        ]
        .drop_duplicates()
        .tolist()
    )

    raise ValueError(
        f"Factory mapping failed.\n"
        f"Unmapped rows: {unmapped_rows}\n"
        f"Products: {unmapped_products}"
    )

print("✓ Factory mapping validated")
print(
    f"✓ Products: "
    f"{df_ml['Product Name'].nunique()}"
)
print(
    f"✓ Factories: "
    f"{df_ml['Factory'].nunique()}"
)
print(
    f"✓ Unmapped rows: "
    f"{unmapped_rows}"
)


# ------------------------------------------------------------
# 8. DATE CONVERSION
# ------------------------------------------------------------

df_ml["Order Date"] = pd.to_datetime(
    df_ml["Order Date"],
    format="%d-%m-%Y",
    errors="raise"
)

df_ml["Ship Date"] = pd.to_datetime(
    df_ml["Ship Date"],
    format="%d-%m-%Y",
    errors="raise"
)

print("✓ Order Date parsed")
print("✓ Ship Date parsed")


# ------------------------------------------------------------
# 9. DELIVERY DAYS
# ------------------------------------------------------------

df_ml["Delivery_Days"] = (
    df_ml["Ship Date"] -
    df_ml["Order Date"]
).dt.days


if df_ml["Delivery_Days"].isna().any():
    raise ValueError(
        "Missing Delivery_Days detected."
    )


if (df_ml["Delivery_Days"] < 0).any():
    raise ValueError(
        "Negative Delivery_Days detected."
    )


print("✓ Delivery_Days calculated")

print(
    f"✓ Mean lead time: "
    f"{df_ml['Delivery_Days'].mean():.2f} days"
)

print(
    f"✓ Median lead time: "
    f"{df_ml['Delivery_Days'].median():.0f} days"
)


# ------------------------------------------------------------
# 10. ORDER-LEVEL TRAIN / TEST SPLIT
# ------------------------------------------------------------

splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)


train_indices, test_indices = next(
    splitter.split(
        df_ml,
        groups=df_ml["Order ID"]
    )
)


train_df = df_ml.iloc[
    train_indices
].copy()

test_df = df_ml.iloc[
    test_indices
].copy()


print("✓ Train/Test split completed")


# ------------------------------------------------------------
# 11. ORDER LEAKAGE CHECK
# ------------------------------------------------------------

train_orders = set(
    train_df["Order ID"].unique()
)

test_orders = set(
    test_df["Order ID"].unique()
)

order_overlap = (
    train_orders
    .intersection(test_orders)
)


if len(order_overlap) != 0:

    raise ValueError(
        f"ORDER LEAKAGE DETECTED: "
        f"{len(order_overlap)} orders overlap."
    )


print("✓ Order ID overlap: 0")
print("✓ No order-level leakage")


# ------------------------------------------------------------
# 12. TRAIN-ONLY Q3 THRESHOLD
# ------------------------------------------------------------

train_q3_threshold = float(
    train_df["Delivery_Days"].quantile(0.75)
)


train_df["Slow_Shipment"] = (
    train_df["Delivery_Days"]
    > train_q3_threshold
).astype(int)


test_df["Slow_Shipment"] = (
    test_df["Delivery_Days"]
    > train_q3_threshold
).astype(int)


print(
    f"✓ Train-only Q3 threshold: "
    f"{train_q3_threshold:.0f} days"
)


# ------------------------------------------------------------
# 13. TIME FEATURES
# ------------------------------------------------------------

for data in [train_df, test_df]:

    data["Order_Year"] = (
        data["Order Date"].dt.year
    )

    data["Order_Month"] = (
        data["Order Date"].dt.month
    )

    data["Order_Quarter"] = (
        data["Order Date"].dt.quarter
    )

    data["Order_DayOfWeek"] = (
        data["Order Date"].dt.dayofweek
    )


# ------------------------------------------------------------
# 14. ML FEATURES
# ------------------------------------------------------------

feature_columns = [

    "Product Name",

    "Division",

    "Country/Region",

    "State/Province",

    "Region",

    "Factory",

    "Ship Mode",

    "Units",

    "Sales",

    "Cost",

    "Gross Profit",

    "Order_Year",

    "Order_Month",

    "Order_Quarter",

    "Order_DayOfWeek"
]


# ------------------------------------------------------------
# 15. CREATE X / y
# ------------------------------------------------------------

X_train = train_df[
    feature_columns
].copy()

X_test = test_df[
    feature_columns
].copy()

y_train = train_df[
    "Slow_Shipment"
].copy()

y_test = test_df[
    "Slow_Shipment"
].copy()


# ------------------------------------------------------------
# 16. LEAKAGE COLUMN CHECK
# ------------------------------------------------------------

forbidden_columns = [
    "Ship Date",
    "Delivery_Days",
    "Slow_Shipment"
]


leakage_columns = [
    column
    for column in forbidden_columns
    if column in X_train.columns
]


if leakage_columns:

    raise ValueError(
        f"LEAKAGE COLUMNS FOUND: "
        f"{leakage_columns}"
    )


print("✓ No target leakage")
print("✓ Delivery_Days not used as feature")
print("✓ Ship Date not used as feature")


# ------------------------------------------------------------
# 17. TARGET VALIDATION
# ------------------------------------------------------------

if y_train.nunique() != 2:
    raise ValueError(
        "Training target does not contain both classes."
    )

if y_test.nunique() != 2:
    raise ValueError(
        "Testing target does not contain both classes."
    )


# ------------------------------------------------------------
# 18. FINAL REPORT
# ------------------------------------------------------------

print()
print("=" * 70)
print("STEP 25A — FINAL VALIDATION REPORT")
print("=" * 70)

print(
    f"Total records          : "
    f"{len(df_ml):,}"
)

print(
    f"Unique orders          : "
    f"{df_ml['Order ID'].nunique():,}"
)

print(
    f"Training records       : "
    f"{len(train_df):,}"
)

print(
    f"Testing records        : "
    f"{len(test_df):,}"
)

print(
    f"Training orders        : "
    f"{train_df['Order ID'].nunique():,}"
)

print(
    f"Testing orders         : "
    f"{test_df['Order ID'].nunique():,}"
)

print(
    f"Order overlap          : "
    f"{len(order_overlap)}"
)

print(
    f"Train Q3 threshold     : "
    f"{train_q3_threshold:.0f} days"
)

print(
    f"Train slow shipments   : "
    f"{y_train.sum():,} "
    f"({y_train.mean() * 100:.2f}%)"
)

print(
    f"Test slow shipments    : "
    f"{y_test.sum():,} "
    f"({y_test.mean() * 100:.2f}%)"
)

print(
    f"Number of features     : "
    f"{len(feature_columns)}"
)

print(
    f"X_train shape          : "
    f"{X_train.shape}"
)

print(
    f"X_test shape           : "
    f"{X_test.shape}"
)

print()
print("✓ scikit-learn available")
print("✓ Factory mapping validated")
print("✓ Dates validated")
print("✓ Delivery_Days validated")
print("✓ Order-level split validated")
print("✓ Order leakage = 0")
print("✓ Target leakage = 0")
print("✓ Train-only threshold used")
print("✓ Both target classes available")
print("✓ ML feature matrix created")

print()
print("STATUS: STEP 25A COMPLETED SUCCESSFULLY")
print("=" * 70)
````

````python
# ============================================================
# STEP 25B — ML MODEL TRAINING & EVALUATION
# ============================================================

import sys
import subprocess
import importlib.util
import os
import re
import unicodedata
import warnings

warnings.filterwarnings("ignore")

# ------------------------------------------------------------
# 1. CHECK / INSTALL REQUIRED LIBRARIES
# ------------------------------------------------------------

required_packages = {
    "pandas": "pandas",
    "numpy": "numpy",
    "sklearn": "scikit-learn"
}

for module_name, package_name in required_packages.items():
    if importlib.util.find_spec(module_name) is None:
        print(f"Installing missing package: {package_name}")
        subprocess.check_call([
            sys.executable,
            "-m",
            "pip",
            "install",
            package_name
        ])

import pandas as pd
import numpy as np

from sklearn.model_selection import GroupShuffleSplit
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

import sklearn

print("=" * 70)
print("STEP 25B — ML MODEL TRAINING & EVALUATION")
print("=" * 70)
print(f"✓ Python environment: {sys.executable}")
print(f"✓ scikit-learn version: {sklearn.__version__}")


# ------------------------------------------------------------
# 2. LOCATE DATASET
# ------------------------------------------------------------

possible_paths = [
    "../data/Nassau Candy Distributor.csv",
    "/mnt/data/Nassau Candy Distributor.csv",
    "/mnt/data/files/Nassau Candy Distributor.csv"
]

data_path = None

for path in possible_paths:
    if os.path.exists(path):
        data_path = path
        break

if data_path is None:
    raise FileNotFoundError(
        "Nassau Candy Distributor.csv was not found."
    )

df_ml = pd.read_csv(data_path)

print(f"✓ Dataset loaded: {data_path}")
print(f"✓ Records: {len(df_ml):,}")


# ------------------------------------------------------------
# 3. REQUIRED COLUMN VALIDATION
# ------------------------------------------------------------

required_columns = [
    "Row ID",
    "Order ID",
    "Order Date",
    "Ship Date",
    "Ship Mode",
    "Customer ID",
    "Country/Region",
    "City",
    "State/Province",
    "Postal Code",
    "Division",
    "Region",
    "Product ID",
    "Product Name",
    "Sales",
    "Units",
    "Gross Profit",
    "Cost"
]

missing_columns = [
    col for col in required_columns
    if col not in df_ml.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}"
    )

print(f"✓ Missing required columns: {missing_columns}")


# ------------------------------------------------------------
# 4. VALIDATED PRODUCT → FACTORY MAPPING
# ------------------------------------------------------------

def canonical_product_name(value):
    if pd.isna(value):
        return ""

    text = str(value)

    text = unicodedata.normalize("NFKC", text)
    text = text.replace("\u00A0", " ")
    text = text.strip()

    text = re.sub(r"\s*-\s*", " - ", text)
    text = re.sub(r"\s+", " ", text)

    return text


product_factory_map = {
    "Wonka Bar - Nutty Crunch Surprise": "Lot's O' Nuts",
    "Wonka Bar - Fudge Mallows": "Lot's O' Nuts",
    "Wonka Bar - Scrumdiddlyumptious": "Lot's O' Nuts",

    "Wonka Bar - Milk Chocolate": "Wicked Choccy's",
    "Wonka Bar - Triple Dazzle Caramel": "Wicked Choccy's",

    "Laffy Taffy": "Sugar Shack",
    "SweeTARTS": "Sugar Shack",
    "Nerds": "Sugar Shack",
    "Fun Dip": "Sugar Shack",
    "Fizzy Lifting Drinks": "Sugar Shack",

    "Everlasting Gobstopper": "Secret Factory",
    "Lickable Wallpaper": "Secret Factory",
    "Wonka Gum": "Secret Factory",

    "Hair Toffee": "The Other Factory",
    "Kazookles": "The Other Factory"
}

df_ml["_Product_Key"] = (
    df_ml["Product Name"]
    .map(canonical_product_name)
)

normalized_map = {
    canonical_product_name(k): v
    for k, v in product_factory_map.items()
}

df_ml["Factory"] = df_ml["_Product_Key"].map(normalized_map)

unmapped_rows = int(df_ml["Factory"].isna().sum())

if unmapped_rows != 0:
    raise ValueError(
        f"Factory mapping failed. Unmapped rows: {unmapped_rows}"
    )

print("✓ Factory mapping validated")
print(f"✓ Products: {df_ml['Product Name'].nunique()}")
print(f"✓ Factories: {df_ml['Factory'].nunique()}")
print(f"✓ Unmapped rows: {unmapped_rows}")


# ------------------------------------------------------------
# 5. DATE PROCESSING
# ------------------------------------------------------------

df_ml["Order Date"] = pd.to_datetime(
    df_ml["Order Date"],
    format="%d-%m-%Y",
    errors="raise"
)

df_ml["Ship Date"] = pd.to_datetime(
    df_ml["Ship Date"],
    format="%d-%m-%Y",
    errors="raise"
)

df_ml["Delivery_Days"] = (
    df_ml["Ship Date"] - df_ml["Order Date"]
).dt.days

if df_ml["Delivery_Days"].isna().any():
    raise ValueError("Missing Delivery_Days detected.")

if (df_ml["Delivery_Days"] < 0).any():
    raise ValueError("Negative Delivery_Days detected.")

print("✓ Dates parsed")
print("✓ Delivery_Days calculated")


# ------------------------------------------------------------
# 6. ORDER-LEVEL TRAIN / TEST SPLIT
# ------------------------------------------------------------

splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, test_idx = next(
    splitter.split(
        df_ml,
        groups=df_ml["Order ID"]
    )
)

train_df = df_ml.iloc[train_idx].copy()
test_df = df_ml.iloc[test_idx].copy()

train_orders = set(train_df["Order ID"])
test_orders = set(test_df["Order ID"])

order_overlap = train_orders.intersection(test_orders)

if len(order_overlap) != 0:
    raise ValueError(
        f"Order leakage detected: {len(order_overlap)} overlapping orders."
    )

print("✓ Order-level train/test split completed")
print(f"✓ Train records: {len(train_df):,}")
print(f"✓ Test records: {len(test_df):,}")
print(f"✓ Train orders: {len(train_orders):,}")
print(f"✓ Test orders: {len(test_orders):,}")
print(f"✓ Order overlap: {len(order_overlap)}")


# ------------------------------------------------------------
# 7. TRAIN-ONLY TARGET THRESHOLD
# ------------------------------------------------------------

train_q3 = float(
    train_df["Delivery_Days"].quantile(0.75)
)

test_q3 = float(
    test_df["Delivery_Days"].quantile(0.75)
)

train_df["Slow_Shipment"] = (
    train_df["Delivery_Days"] > train_q3
).astype(int)

test_df["Slow_Shipment"] = (
    test_df["Delivery_Days"] > train_q3
).astype(int)

print(f"✓ Train-only Q3 threshold: {train_q3:.0f} days")
print(
    f"✓ Train slow shipments: "
    f"{train_df['Slow_Shipment'].sum():,}"
)
print(
    f"✓ Test slow shipments: "
    f"{test_df['Slow_Shipment'].sum():,}"
)


# ------------------------------------------------------------
# 8. TIME FEATURES
# ------------------------------------------------------------

for data in [train_df, test_df]:

    data["Order_Year"] = data["Order Date"].dt.year
    data["Order_Month"] = data["Order Date"].dt.month
    data["Order_Quarter"] = data["Order Date"].dt.quarter
    data["Order_DayOfWeek"] = data["Order Date"].dt.dayofweek


# ------------------------------------------------------------
# 9. DEFINE ML FEATURES
# ------------------------------------------------------------

categorical_features = [
    "Product Name",
    "Division",
    "Country/Region",
    "State/Province",
    "Region",
    "Factory",
    "Ship Mode"
]

numeric_features = [
    "Units",
    "Sales",
    "Cost",
    "Gross Profit",
    "Order_Year",
    "Order_Month",
    "Order_Quarter",
    "Order_DayOfWeek"
]

feature_columns = (
    categorical_features +
    numeric_features
)

target_column = "Slow_Shipment"

X_train = train_df[feature_columns].copy()
X_test = test_df[feature_columns].copy()

y_train = train_df[target_column].copy()
y_test = test_df[target_column].copy()


# ------------------------------------------------------------
# 10. LEAKAGE CHECK
# ------------------------------------------------------------

forbidden_columns = [
    "Ship Date",
    "Delivery_Days",
    "Slow_Shipment"
]

leakage_columns = [
    col for col in forbidden_columns
    if col in feature_columns
]

if leakage_columns:
    raise ValueError(
        f"Data leakage detected in features: {leakage_columns}"
    )

print(f"✓ ML features: {len(feature_columns)}")
print(f"✓ Leakage columns detected: {leakage_columns}")


# ------------------------------------------------------------
# 11. TARGET VALIDATION
# ------------------------------------------------------------

if y_train.nunique() != 2:
    raise ValueError(
        "Training target does not contain both classes."
    )

if y_test.nunique() != 2:
    raise ValueError(
        "Testing target does not contain both classes."
    )

print(
    f"✓ Train target classes: "
    f"{sorted(y_train.unique().tolist())}"
)

print(
    f"✓ Test target classes: "
    f"{sorted(y_test.unique().tolist())}"
)


# ------------------------------------------------------------
# 12. ONE-HOT ENCODER COMPATIBILITY
# ------------------------------------------------------------

import inspect

ohe_parameters = {
    "handle_unknown": "ignore"
}

if "sparse_output" in inspect.signature(
    OneHotEncoder
).parameters:

    ohe_parameters["sparse_output"] = False

else:

    ohe_parameters["sparse"] = False


# ------------------------------------------------------------
# 13. PREPROCESSING PIPELINES
# ------------------------------------------------------------

numeric_transformer_scaled = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)

numeric_transformer_tree = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        )
    ]
)

categorical_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "onehot",
            OneHotEncoder(**ohe_parameters)
        )
    ]
)


# ------------------------------------------------------------
# 14. LOGISTIC REGRESSION PREPROCESSOR
# ------------------------------------------------------------

preprocessor_logistic = ColumnTransformer(
    transformers=[
        (
            "num",
            numeric_transformer_scaled,
            numeric_features
        ),
        (
            "cat",
            categorical_transformer,
            categorical_features
        )
    ],
    remainder="drop"
)


# ------------------------------------------------------------
# 15. TREE MODEL PREPROCESSOR
# ------------------------------------------------------------

preprocessor_tree = ColumnTransformer(
    transformers=[
        (
            "num",
            numeric_transformer_tree,
            numeric_features
        ),
        (
            "cat",
            categorical_transformer,
            categorical_features
        )
    ],
    remainder="drop"
)


# ------------------------------------------------------------
# 16. DEFINE MODELS
# ------------------------------------------------------------

models = {

    "Logistic Regression": Pipeline(
        steps=[
            (
                "preprocessor",
                preprocessor_logistic
            ),
            (
                "model",
                LogisticRegression(
                    max_iter=2000,
                    class_weight="balanced",
                    random_state=42
                )
            )
        ]
    ),

    "Random Forest": Pipeline(
        steps=[
            (
                "preprocessor",
                preprocessor_tree
            ),
            (
                "model",
                RandomForestClassifier(
                    n_estimators=300,
                    max_depth=None,
                    min_samples_split=5,
                    min_samples_leaf=2,
                    class_weight="balanced",
                    random_state=42,
                    n_jobs=-1
                )
            )
        ]
    ),

    "Gradient Boosting": Pipeline(
        steps=[
            (
                "preprocessor",
                preprocessor_tree
            ),
            (
                "model",
                GradientBoostingClassifier(
                    n_estimators=200,
                    learning_rate=0.05,
                    max_depth=3,
                    random_state=42
                )
            )
        ]
    )
}


# ------------------------------------------------------------
# 17. TRAIN + EVALUATE MODELS
# ------------------------------------------------------------

results = []
confusion_matrices = {}

for model_name, model in models.items():

    print("\n" + "-" * 70)
    print(f"TRAINING: {model_name}")
    print("-" * 70)

    # Train ONLY on training data
    model.fit(X_train, y_train)

    # Predictions on unseen test data
    y_pred = model.predict(X_test)

    # Probability for ROC-AUC
    y_prob = model.predict_proba(X_test)[:, 1]

    accuracy = accuracy_score(
        y_test,
        y_pred
    )

    precision = precision_score(
        y_test,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_test,
        y_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        y_pred,
        zero_division=0
    )

    roc_auc = roc_auc_score(
        y_test,
        y_prob
    )

    cm = confusion_matrix(
        y_test,
        y_pred
    )

    confusion_matrices[model_name] = cm

    results.append({
        "Model": model_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1_Score": f1,
        "ROC_AUC": roc_auc
    })

    print(f"Accuracy : {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall   : {recall:.4f}")
    print(f"F1 Score : {f1:.4f}")
    print(f"ROC-AUC  : {roc_auc:.4f}")

    print("\nConfusion Matrix:")
    print(cm)


# ------------------------------------------------------------
# 18. MODEL COMPARISON
# ------------------------------------------------------------

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="ROC_AUC",
    ascending=False
).reset_index(drop=True)

print("\n" + "=" * 70)
print("MODEL COMPARISON")
print("=" * 70)

print(
    results_df.to_string(
        index=False,
        formatters={
            "Accuracy": "{:.4f}".format,
            "Precision": "{:.4f}".format,
            "Recall": "{:.4f}".format,
            "F1_Score": "{:.4f}".format,
            "ROC_AUC": "{:.4f}".format
        }
    )
)


# ------------------------------------------------------------
# 19. BEST MODEL
# ------------------------------------------------------------

best_model_name = results_df.iloc[0]["Model"]
best_model = models[best_model_name]

print("\n" + "=" * 70)
print("BEST MODEL")
print("=" * 70)

print(f"✓ Best model based on ROC-AUC: {best_model_name}")
print(
    f"✓ ROC-AUC: "
    f"{results_df.iloc[0]['ROC_AUC']:.4f}"
)


# ------------------------------------------------------------
# 20. FINAL VALIDATION
# ------------------------------------------------------------

assert len(X_train) == len(y_train)
assert len(X_test) == len(y_test)

assert len(order_overlap) == 0

assert not leakage_columns

assert len(results_df) == 3

assert results_df["ROC_AUC"].between(
    0,
    1
).all()

print("\n" + "=" * 70)
print("STATUS: STEP 25B COMPLETED SUCCESSFULLY")
print("=" * 70)
````

````python
# ============================================================
# STEP 26A — FACTORY REALLOCATION SCENARIO ENGINE
# ============================================================

import os
import re
import sys
import subprocess
import unicodedata
from pathlib import Path

import numpy as np
import pandas as pd

print("=" * 75)
print("STEP 26A — FACTORY REALLOCATION SCENARIO ENGINE")
print("=" * 75)

# ------------------------------------------------------------
# 1. PROJECT PATHS
# ------------------------------------------------------------

cwd = Path.cwd()

if cwd.name.lower() == "notebooks":
    project_root = cwd.parent
else:
    project_root = cwd

data_path = project_root / "data" / "Nassau Candy Distributor.csv"
models_dir = project_root / "models"
model_path = models_dir / "nassau_slow_shipment_model.joblib"

if not data_path.exists():
    raise FileNotFoundError(
        f"Dataset not found:\n{data_path}"
    )

if not model_path.exists():
    raise FileNotFoundError(
        f"Saved model not found:\n{model_path}\n"
        "Please complete Step 25D first."
    )

print("✓ Project root:", project_root)
print("✓ Dataset found")
print("✓ Saved model found")

# ------------------------------------------------------------
# 2. LOAD SAVED MODEL
# ------------------------------------------------------------

try:
    import joblib
except ImportError:
    subprocess.check_call(
        [sys.executable, "-m", "pip", "install", "joblib", "-q"]
    )
    import joblib

scenario_model = joblib.load(model_path)

if scenario_model is None:
    raise RuntimeError(
        "Saved model could not be loaded."
    )

print("✓ Random Forest model loaded")

# ------------------------------------------------------------
# 3. LOAD RAW DATA
# ------------------------------------------------------------

raw_df = pd.read_csv(data_path)

if raw_df.empty:
    raise RuntimeError("Dataset is empty.")

print("✓ Dataset loaded")
print("  Records:", len(raw_df))
print("  Columns:", len(raw_df.columns))

# ------------------------------------------------------------
# 4. REQUIRED RAW COLUMNS
# ------------------------------------------------------------

required_raw_columns = [
    "Order ID",
    "Order Date",
    "Ship Date",
    "Ship Mode",
    "Customer ID",
    "Country/Region",
    "State/Province",
    "Region",
    "Product Name",
    "Division",
    "Units",
    "Sales",
    "Cost",
    "Gross Profit"
]

missing_raw_columns = [
    col for col in required_raw_columns
    if col not in raw_df.columns
]

if missing_raw_columns:
    raise RuntimeError(
        f"Missing raw columns: {missing_raw_columns}"
    )

print("✓ Required raw columns verified")

# ------------------------------------------------------------
# 5. PRODUCT NAME CANONICALIZATION
# ------------------------------------------------------------

def canonical_product_name(value):
    if pd.isna(value):
        return ""

    text = str(value)

    text = unicodedata.normalize(
        "NFKC",
        text
    )

    text = text.replace(
        "\u00A0",
        " "
    )

    text = text.strip()

    text = re.sub(
        r"\s*-\s*",
        " - ",
        text
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    )

    return text

# ------------------------------------------------------------
# 6. VALIDATED PRODUCT → FACTORY MAPPING
# ------------------------------------------------------------

product_to_factory = {

    # Lot's O' Nuts
    "Wonka Bar - Nutty Crunch Surprise": "Lot's O' Nuts",
    "Wonka Bar - Fudge Mallows": "Lot's O' Nuts",
    "Wonka Bar - Scrumdiddlyumptious": "Lot's O' Nuts",

    # Wicked Choccy's
    "Wonka Bar - Milk Chocolate": "Wicked Choccy's",
    "Wonka Bar - Triple Dazzle Caramel": "Wicked Choccy's",

    # Sugar Shack
    "Laffy Taffy": "Sugar Shack",
    "SweeTARTS": "Sugar Shack",
    "Nerds": "Sugar Shack",
    "Fun Dip": "Sugar Shack",
    "Fizzy Lifting Drinks": "Sugar Shack",

    # Secret Factory
    "Everlasting Gobstopper": "Secret Factory",
    "Lickable Wallpaper": "Secret Factory",
    "Wonka Gum": "Secret Factory",

    # The Other Factory
    "Hair Toffee": "The Other Factory",
    "Kazookles": "The Other Factory"
}

# Normalize mapping keys
normalized_product_to_factory = {
    canonical_product_name(product): factory
    for product, factory in product_to_factory.items()
}

raw_df["_Product_Key"] = raw_df["Product Name"].apply(
    canonical_product_name
)

raw_df["Factory"] = raw_df["_Product_Key"].map(
    normalized_product_to_factory
)

# ------------------------------------------------------------
# 7. FACTORY MAPPING VALIDATION
# ------------------------------------------------------------

unmapped_rows = int(
    raw_df["Factory"].isna().sum()
)

if unmapped_rows != 0:
    unmapped_products = sorted(
        raw_df.loc[
            raw_df["Factory"].isna(),
            "Product Name"
        ].astype(str).unique()
    )

    raise RuntimeError(
        "Factory mapping failed.\n"
        f"Unmapped rows: {unmapped_rows}\n"
        f"Unmapped products: {unmapped_products}"
    )

factories = sorted(
    raw_df["Factory"].dropna().unique().tolist()
)

if len(factories) != 5:
    raise RuntimeError(
        f"Expected 5 factories, found {len(factories)}: {factories}"
    )

product_count = raw_df["Product Name"].nunique()

if product_count != 15:
    raise RuntimeError(
        f"Expected 15 products, found {product_count}"
    )

print("✓ Factory mapping validated")
print("  Products :", product_count)
print("  Factories:", len(factories))
print("  Unmapped :", unmapped_rows)

# ------------------------------------------------------------
# 8. DATE FEATURES
# ------------------------------------------------------------

raw_df["Order Date"] = pd.to_datetime(
    raw_df["Order Date"],
    format="%d-%m-%Y",
    errors="raise"
)

raw_df["Ship Date"] = pd.to_datetime(
    raw_df["Ship Date"],
    format="%d-%m-%Y",
    errors="raise"
)

delivery_days = (
    raw_df["Ship Date"] -
    raw_df["Order Date"]
).dt.days

if delivery_days.isna().any():
    raise RuntimeError(
        "Delivery_Days contains missing values."
    )

if (delivery_days < 0).any():
    raise RuntimeError(
        "Negative Delivery_Days detected."
    )

raw_df["Delivery_Days"] = delivery_days

raw_df["Order_Year"] = raw_df["Order Date"].dt.year
raw_df["Order_Month"] = raw_df["Order Date"].dt.month
raw_df["Order_Quarter"] = raw_df["Order Date"].dt.quarter
raw_df["Order_DayOfWeek"] = raw_df["Order Date"].dt.dayofweek

print("✓ Date features recreated")

# ------------------------------------------------------------
# 9. MODEL FEATURE DEFINITION
# ------------------------------------------------------------

feature_columns = [
    "Product Name",
    "Division",
    "Country/Region",
    "State/Province",
    "Region",
    "Factory",
    "Ship Mode",
    "Units",
    "Sales",
    "Cost",
    "Gross Profit",
    "Order_Year",
    "Order_Month",
    "Order_Quarter",
    "Order_DayOfWeek"
]

missing_features = [
    col for col in feature_columns
    if col not in raw_df.columns
]

if missing_features:
    raise RuntimeError(
        f"Missing model features: {missing_features}"
    )

# ------------------------------------------------------------
# 10. BUILD BASE FEATURE DATASET
# ------------------------------------------------------------

base_features = raw_df[
    feature_columns
].copy()

if base_features.isna().any().any():
    missing_feature_counts = (
        base_features.isna().sum()
    )

    raise RuntimeError(
        "Missing values detected in model features:\n"
        f"{missing_feature_counts[missing_feature_counts > 0]}"
    )

print("✓ Model feature dataset created")
print("  Features:", len(feature_columns))

# ------------------------------------------------------------
# 11. CURRENT FACTORY BASELINE
# ------------------------------------------------------------

current_factory_by_product = (
    raw_df[
        ["Product Name", "Factory"]
    ]
    .drop_duplicates()
)

factory_counts_per_product = (
    current_factory_by_product
    .groupby("Product Name")["Factory"]
    .nunique()
)

if (factory_counts_per_product != 1).any():
    problematic_products = (
        factory_counts_per_product[
            factory_counts_per_product != 1
        ]
    )

    raise RuntimeError(
        "Some products have multiple current factories:\n"
        f"{problematic_products}"
    )

print("✓ Current factory baseline validated")

# ------------------------------------------------------------
# 12. GENERATE COUNTERFACTUAL SCENARIOS
# ------------------------------------------------------------
#
# For every product:
#   Current Factory
#   Alternative Factory 1
#   Alternative Factory 2
#   ...
#
# All other operational features remain unchanged.
#
# This is scenario analysis, not causal inference.
# ------------------------------------------------------------

scenario_results = []

unique_products = sorted(
    raw_df["Product Name"].dropna().unique()
)

for product in unique_products:

    product_mask = (
        raw_df["Product Name"] == product
    )

    product_data = base_features.loc[
        product_mask
    ].copy()

    current_factory = (
        raw_df.loc[
            product_mask,
            "Factory"
        ]
        .iloc[0]
    )

    for scenario_factory in factories:

        scenario_data = product_data.copy()

        scenario_data["Factory"] = scenario_factory

        # Predict probability of Slow Shipment = class 1
        probabilities = scenario_model.predict_proba(
            scenario_data
        )[:, 1]

        predictions = scenario_model.predict(
            scenario_data
        )

        scenario_results.append({
            "Product Name": product,
            "Current Factory": current_factory,
            "Scenario Factory": scenario_factory,
            "Records Scored": len(scenario_data),
            "Predicted Slow Probability": float(
                np.mean(probabilities)
            ),
            "Predicted Slow Rate": float(
                np.mean(predictions)
            )
        })

scenario_df = pd.DataFrame(
    scenario_results
)

# ------------------------------------------------------------
# 13. VALIDATE SCENARIO DATASET
# ------------------------------------------------------------

expected_rows = (
    len(unique_products) *
    len(factories)
)

if len(scenario_df) != expected_rows:
    raise RuntimeError(
        f"Expected {expected_rows} scenario rows, "
        f"found {len(scenario_df)}."
    )

if scenario_df["Predicted Slow Probability"].isna().any():
    raise RuntimeError(
        "Scenario predictions contain missing values."
    )

if not (
    scenario_df["Predicted Slow Probability"]
    .between(0, 1)
    .all()
):
    raise RuntimeError(
        "Predicted probabilities outside [0, 1]."
    )

print("✓ Scenario predictions generated")
print("  Products :", len(unique_products))
print("  Factories:", len(factories))
print("  Scenarios:", len(scenario_df))

# ------------------------------------------------------------
# 14. CURRENT FACTORY BASELINE PROBABILITY
# ------------------------------------------------------------

current_baseline = (
    scenario_df[
        scenario_df["Scenario Factory"]
        ==
        scenario_df["Current Factory"]
    ]
    [
        [
            "Product Name",
            "Current Factory",
            "Predicted Slow Probability"
        ]
    ]
    .rename(
        columns={
            "Predicted Slow Probability":
            "Current Slow Probability"
        }
    )
)

# ------------------------------------------------------------
# 15. CALCULATE REALLOCATION IMPROVEMENT
# ------------------------------------------------------------

scenario_df = scenario_df.merge(
    current_baseline,
    on=[
        "Product Name",
        "Current Factory"
    ],
    how="left",
    validate="many_to_one"
)

scenario_df["Probability Reduction"] = (
    scenario_df["Current Slow Probability"]
    -
    scenario_df["Predicted Slow Probability"]
)

scenario_df["Improvement_Percentage_Points"] = (
    scenario_df["Probability Reduction"] * 100
)

scenario_df["Is_Current_Factory"] = (
    scenario_df["Scenario Factory"]
    ==
    scenario_df["Current Factory"]
)

# ------------------------------------------------------------
# 16. ALTERNATIVE FACTORY RECOMMENDATION
# ------------------------------------------------------------

alternative_scenarios = (
    scenario_df[
        ~scenario_df["Is_Current_Factory"]
    ]
    .copy()
)

best_alternatives = (
    alternative_scenarios
    .sort_values(
        [
            "Product Name",
            "Probability Reduction",
            "Predicted Slow Probability"
        ],
        ascending=[
            True,
            False,
            True
        ]
    )
    .groupby(
        "Product Name",
        as_index=False
    )
    .head(1)
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# 17. ADD RECOMMENDATION STATUS
# ------------------------------------------------------------
#
# We use a conservative screening threshold:
# at least 5 percentage-point reduction in predicted
# slow-shipment probability.
#
# This is NOT a causal threshold. It is only a
# scenario-screening rule.
# ------------------------------------------------------------

recommendation_threshold = 0.05

best_alternatives["Recommendation"] = np.where(
    best_alternatives["Probability Reduction"]
    >= recommendation_threshold,
    "CONSIDER REALLOCATION",
    "NO STRONG ML SIGNAL"
)

# ------------------------------------------------------------
# 18. FINAL RECOMMENDATION TABLE
# ------------------------------------------------------------

recommendation_columns = [
    "Product Name",
    "Current Factory",
    "Scenario Factory",
    "Current Slow Probability",
    "Predicted Slow Probability",
    "Probability Reduction",
    "Improvement_Percentage_Points",
    "Recommendation"
]

recommendation_df = (
    best_alternatives[
        recommendation_columns
    ]
    .sort_values(
        "Probability Reduction",
        ascending=False
    )
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# 19. SAVE RESULTS
# ------------------------------------------------------------

results_dir = project_root / "results"
results_dir.mkdir(
    parents=True,
    exist_ok=True
)

scenario_path = (
    results_dir /
    "factory_reallocation_scenarios.csv"
)

recommendation_path = (
    results_dir /
    "factory_reallocation_recommendations.csv"
)

scenario_df.to_csv(
    scenario_path,
    index=False
)

recommendation_df.to_csv(
    recommendation_path,
    index=False
)

# ------------------------------------------------------------
# 20. DISPLAY RESULTS
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("TOP FACTORY REALLOCATION SCENARIOS")
print("=" * 75)

display(
    recommendation_df[
        [
            "Product Name",
            "Current Factory",
            "Scenario Factory",
            "Current Slow Probability",
            "Predicted Slow Probability",
            "Improvement_Percentage_Points",
            "Recommendation"
        ]
    ].head(15)
)

# ------------------------------------------------------------
# 21. FINAL VALIDATION
# ------------------------------------------------------------

if recommendation_df["Product Name"].nunique() != 15:
    raise RuntimeError(
        "Expected one best alternative for each of 15 products."
    )

if recommendation_df["Scenario Factory"].isna().any():
    raise RuntimeError(
        "Missing recommended scenario factory."
    )

if (
    recommendation_df["Scenario Factory"]
    ==
    recommendation_df["Current Factory"]
).any():
    raise RuntimeError(
        "Current factory incorrectly selected as alternative."
    )

print("\n" + "=" * 75)
print("STATUS: STEP 26A COMPLETED SUCCESSFULLY")
print("=" * 75)

print("Total products evaluated :", len(unique_products))
print("Factories evaluated      :", len(factories))
print("Scenario combinations    :", len(scenario_df))
print("Recommendations generated:", len(recommendation_df))
print("Scenario file            :", scenario_path.name)
print("Recommendation file      :", recommendation_path.name)

print("\nIMPORTANT:")
print("These are ML-based counterfactual scenario recommendations.")
print("They are NOT causal estimates and do NOT use invented factory coordinates.")

print("=" * 75)
````

````python
# ============================================================
# STEP 27 — RANDOM FOREST MODEL EXPLAINABILITY
# Feature Importance Analysis
# ============================================================

from pathlib import Path
import sys
import subprocess
import importlib.util

# ------------------------------------------------------------
# 1. Required package check
# ------------------------------------------------------------
if importlib.util.find_spec("joblib") is None:
    subprocess.check_call([sys.executable, "-m", "pip", "install", "joblib"])

import joblib
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 2. Locate project directory and saved model
# ------------------------------------------------------------
current_path = Path.cwd()

possible_model_paths = [
    current_path / "models" / "nassau_slow_shipment_model.joblib",
    current_path.parent / "models" / "nassau_slow_shipment_model.joblib",
]

model_path = next(
    (p for p in possible_model_paths if p.exists()),
    None
)

if model_path is None:
    raise FileNotFoundError(
        "Saved model not found. Expected:\n"
        "models/nassau_slow_shipment_model.joblib"
    )

# ------------------------------------------------------------
# 3. Load saved model
# ------------------------------------------------------------
loaded_model = joblib.load(model_path)

print("=" * 70)
print("STEP 27 — RANDOM FOREST FEATURE IMPORTANCE")
print("=" * 70)

print(f"Model file: {model_path}")
print(f"Loaded model type: {type(loaded_model).__name__}")

# ------------------------------------------------------------
# 4. Validate that the saved object is a Pipeline
# ------------------------------------------------------------
if not hasattr(loaded_model, "named_steps"):
    raise TypeError(
        "Saved model is not a scikit-learn Pipeline. "
        "Feature extraction cannot continue safely."
    )

print(f"Pipeline steps: {list(loaded_model.named_steps.keys())}")

# ------------------------------------------------------------
# 5. Find the fitted estimator
# ------------------------------------------------------------
estimator = None

for step_name, step_object in reversed(loaded_model.named_steps.items()):
    if hasattr(step_object, "feature_importances_"):
        estimator = step_object
        estimator_name = step_name
        break

if estimator is None:
    raise TypeError(
        "No estimator with feature_importances_ was found "
        "inside the saved pipeline."
    )

print(f"Estimator step: {estimator_name}")
print(f"Estimator type: {type(estimator).__name__}")

# ------------------------------------------------------------
# 6. Confirm Random Forest
# ------------------------------------------------------------
if not hasattr(estimator, "feature_importances_"):
    raise AttributeError(
        "The selected estimator does not provide feature_importances_."
    )

raw_importances = np.asarray(estimator.feature_importances_, dtype=float)

if len(raw_importances) == 0:
    raise ValueError("Feature importance array is empty.")

if not np.isfinite(raw_importances).all():
    raise ValueError("Feature importance contains invalid numeric values.")

# ------------------------------------------------------------
# 7. Locate preprocessing transformer
# ------------------------------------------------------------
preprocessor = None

for step_name, step_object in loaded_model.named_steps.items():
    if hasattr(step_object, "get_feature_names_out"):
        if step_name != estimator_name:
            preprocessor = step_object
            preprocessor_name = step_name
            break

if preprocessor is None:
    raise TypeError(
        "Preprocessing transformer with get_feature_names_out() "
        "was not found in the saved pipeline."
    )

print(f"Preprocessor step: {preprocessor_name}")

# ------------------------------------------------------------
# 8. Extract transformed feature names
# ------------------------------------------------------------
feature_names = np.asarray(
    preprocessor.get_feature_names_out(),
    dtype=str
)

if len(feature_names) != len(raw_importances):
    raise ValueError(
        f"Feature-name count ({len(feature_names)}) does not match "
        f"importance count ({len(raw_importances)})."
    )

# ------------------------------------------------------------
# 9. Create transformed-feature importance table
# ------------------------------------------------------------
importance_df = pd.DataFrame({
    "Transformed_Feature": feature_names,
    "Importance": raw_importances
})

importance_df["Importance_Percentage"] = (
    importance_df["Importance"] /
    importance_df["Importance"].sum()
) * 100

importance_df = importance_df.sort_values(
    "Importance",
    ascending=False
).reset_index(drop=True)

# ------------------------------------------------------------
# 10. Aggregate One-Hot Encoded features back to original features
# ------------------------------------------------------------
original_features = [
    "Product Name",
    "Division",
    "Country/Region",
    "State/Province",
    "Region",
    "Factory",
    "Ship Mode",
    "Units",
    "Sales",
    "Cost",
    "Gross Profit",
    "Order_Year",
    "Order_Month",
    "Order_Quarter",
    "Order_DayOfWeek",
]

def map_to_original_feature(transformed_name):
    """
    Convert ColumnTransformer output names such as:
    cat__Product Name_X
    num__Sales
    back to the original feature name.
    """
    
    cleaned = str(transformed_name)

    if cleaned.startswith("cat__"):
        cleaned = cleaned[len("cat__"):]

        for feature in sorted(
            original_features,
            key=len,
            reverse=True
        ):
            if cleaned == feature or cleaned.startswith(feature + "_"):
                return feature

    if cleaned.startswith("num__"):
        cleaned = cleaned[len("num__"):]

        if cleaned in original_features:
            return cleaned

    # Fallback: try direct matching
    for feature in sorted(
        original_features,
        key=len,
        reverse=True
    ):
        if cleaned == feature or cleaned.startswith(feature + "_"):
            return feature

    return cleaned


importance_df["Original_Feature"] = (
    importance_df["Transformed_Feature"]
    .apply(map_to_original_feature)
)

# ------------------------------------------------------------
# 11. Aggregate importance by original feature
# ------------------------------------------------------------
feature_importance_summary = (
    importance_df
    .groupby("Original_Feature", as_index=False)["Importance"]
    .sum()
    .sort_values("Importance", ascending=False)
    .reset_index(drop=True)
)

feature_importance_summary["Importance_Percentage"] = (
    feature_importance_summary["Importance"] /
    feature_importance_summary["Importance"].sum()
) * 100

# ------------------------------------------------------------
# 12. Validation
# ------------------------------------------------------------
if len(feature_importance_summary) != len(original_features):
    missing_features = sorted(
        set(original_features)
        - set(feature_importance_summary["Original_Feature"])
    )

    if missing_features:
        print(
            "Note: Some original features have no transformed "
            f"importance entry: {missing_features}"
        )

if not np.isclose(
    feature_importance_summary["Importance"].sum(),
    1.0,
    atol=1e-6
):
    raise ValueError(
        "Aggregated feature importance does not sum to 1."
    )

if (
    feature_importance_summary["Importance_Percentage"].isna().any()
    or
    np.isinf(
        feature_importance_summary["Importance_Percentage"]
    ).any()
):
    raise ValueError(
        "Invalid values found in feature importance percentages."
    )

# ------------------------------------------------------------
# 13. Save results
# ------------------------------------------------------------
project_dir = model_path.parent.parent
results_dir = project_dir / "results"
results_dir.mkdir(parents=True, exist_ok=True)

transformed_output = (
    results_dir / "random_forest_transformed_feature_importance.csv"
)

summary_output = (
    results_dir / "random_forest_feature_importance.csv"
)

importance_df.to_csv(
    transformed_output,
    index=False
)

feature_importance_summary.to_csv(
    summary_output,
    index=False
)

# ------------------------------------------------------------
# 14. Display top 15 original features
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("TOP MODEL FEATURES")
print("=" * 70)

display_columns = [
    "Original_Feature",
    "Importance",
    "Importance_Percentage"
]

print(
    feature_importance_summary[
        display_columns
    ]
    .head(15)
    .to_string(index=False)
)

# ------------------------------------------------------------
# 15. Final validation summary
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("STEP 27 VALIDATION")
print("=" * 70)

print(f"Transformed features analyzed: {len(feature_names)}")
print(
    f"Original model features analyzed: "
    f"{len(feature_importance_summary)}"
)
print(
    "Importance total: "
    f"{feature_importance_summary['Importance'].sum():.6f}"
)

print(f"\nSaved transformed feature importance:")
print(transformed_output)

print(f"\nSaved original feature importance:")
print(summary_output)

print("\n" + "=" * 70)
print("STATUS: STEP 27 COMPLETED SUCCESSFULLY")
print("=" * 70)
````

````python
# ============================================================
# STEP 28 — TEST-SET PERMUTATION IMPORTANCE
# Independent Model Explainability Validation
# ============================================================

from pathlib import Path
import sys
import subprocess
import importlib.util

# ------------------------------------------------------------
# 1. Check required packages
# ------------------------------------------------------------
required_packages = {
    "pandas": "pandas",
    "numpy": "numpy",
    "joblib": "joblib",
    "sklearn": "scikit-learn"
}

for module_name, package_name in required_packages.items():
    if importlib.util.find_spec(module_name) is None:
        subprocess.check_call(
            [sys.executable, "-m", "pip", "install", package_name]
        )

import pandas as pd
import numpy as np
import joblib

from sklearn.model_selection import GroupShuffleSplit
from sklearn.inspection import permutation_importance
from sklearn.metrics import roc_auc_score


# ============================================================
# 2. Locate project directory
# ============================================================

current_path = Path.cwd()

possible_project_dirs = [
    current_path,
    current_path.parent,
]

project_dir = None

for candidate in possible_project_dirs:
    if (
        (candidate / "data" / "Nassau Candy Distributor.csv").exists()
        and
        (candidate / "models" / "nassau_slow_shipment_model.joblib").exists()
    ):
        project_dir = candidate
        break

if project_dir is None:
    raise FileNotFoundError(
        "Project files could not be located.\n"
        "Expected data/Nassau Candy Distributor.csv and "
        "models/nassau_slow_shipment_model.joblib"
    )


# ============================================================
# 3. Define file paths
# ============================================================

data_path = project_dir / "data" / "Nassau Candy Distributor.csv"
model_path = project_dir / "models" / "nassau_slow_shipment_model.joblib"
results_dir = project_dir / "results"

results_dir.mkdir(parents=True, exist_ok=True)


# ============================================================
# 4. Load raw dataset
# ============================================================

df = pd.read_csv(data_path)

print("=" * 70)
print("STEP 28 — TEST-SET PERMUTATION IMPORTANCE")
print("=" * 70)

print(f"Dataset: {data_path}")
print(f"Records loaded: {len(df):,}")


# ============================================================
# 5. Validate raw dataset
# ============================================================

required_raw_columns = [
    "Row ID",
    "Order ID",
    "Order Date",
    "Ship Date",
    "Ship Mode",
    "Customer ID",
    "Country/Region",
    "City",
    "State/Province",
    "Postal Code",
    "Division",
    "Region",
    "Product ID",
    "Product Name",
    "Sales",
    "Units",
    "Gross Profit",
    "Cost"
]

missing_raw_columns = [
    col for col in required_raw_columns
    if col not in df.columns
]

if missing_raw_columns:
    raise ValueError(
        f"Missing raw dataset columns: {missing_raw_columns}"
    )

if len(df) != 10194:
    raise ValueError(
        f"Unexpected dataset size: {len(df)}. "
        "Expected 10,194 records."
    )


# ============================================================
# 6. Validated Product → Factory mapping
# ============================================================

factory_mapping = {
    "Wonka Bar - Nutty Crunch Surprise": "Lot's O' Nuts",
    "Wonka Bar - Fudge Mallows": "Lot's O' Nuts",
    "Wonka Bar - Scrumdiddlyumptious": "Lot's O' Nuts",

    "Wonka Bar - Milk Chocolate": "Wicked Choccy's",
    "Wonka Bar - Triple Dazzle Caramel": "Wicked Choccy's",

    "Laffy Taffy": "Sugar Shack",
    "SweeTARTS": "Sugar Shack",
    "Nerds": "Sugar Shack",
    "Fun Dip": "Sugar Shack",
    "Fizzy Lifting Drinks": "Sugar Shack",

    "Everlasting Gobstopper": "Secret Factory",
    "Lickable Wallpaper": "Secret Factory",
    "Wonka Gum": "Secret Factory",

    "Hair Toffee": "The Other Factory",
    "Kazookles": "The Other Factory"
}


# ============================================================
# 7. Normalize product names safely
# ============================================================

import re
import unicodedata


def canonical_product_name(value):
    if pd.isna(value):
        return ""

    text = str(value)

    text = unicodedata.normalize("NFKC", text)
    text = text.replace("\u00A0", " ")
    text = text.strip()

    # Normalize spaces around hyphen
    text = re.sub(r"\s*-\s*", " - ", text)

    # Normalize repeated whitespace
    text = re.sub(r"\s+", " ", text)

    return text


df["_Product_Key"] = df["Product Name"].apply(
    canonical_product_name
)

mapping_normalized = {
    canonical_product_name(product): factory
    for product, factory in factory_mapping.items()
}

df["Factory"] = df["_Product_Key"].map(
    mapping_normalized
)

unmapped_rows = int(df["Factory"].isna().sum())

if unmapped_rows != 0:
    raise ValueError(
        f"Factory mapping failed for {unmapped_rows} rows."
    )


# ============================================================
# 8. Parse dates from ORIGINAL raw strings
# ============================================================

df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    format="%d-%m-%Y",
    errors="raise"
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"],
    format="%d-%m-%Y",
    errors="raise"
)

df["Delivery_Days"] = (
    df["Ship Date"] - df["Order Date"]
).dt.days


# ============================================================
# 9. Validate delivery calculations
# ============================================================

if df["Order Date"].isna().any():
    raise ValueError("Order Date contains missing values.")

if df["Ship Date"].isna().any():
    raise ValueError("Ship Date contains missing values.")

if df["Delivery_Days"].isna().any():
    raise ValueError("Delivery_Days contains missing values.")

if (df["Delivery_Days"] < 0).any():
    raise ValueError(
        "Negative Delivery_Days detected."
    )


# ============================================================
# 10. Create date features
# ============================================================

df["Order_Year"] = df["Order Date"].dt.year
df["Order_Month"] = df["Order Date"].dt.month
df["Order_Quarter"] = df["Order Date"].dt.quarter
df["Order_DayOfWeek"] = df["Order Date"].dt.dayofweek


# ============================================================
# 11. Reproduce the SAME order-level split used previously
# ============================================================

splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, test_idx = next(
    splitter.split(
        df,
        groups=df["Order ID"]
    )
)

train_df = df.iloc[train_idx].copy()
test_df = df.iloc[test_idx].copy()


# ============================================================
# 12. Validate order-level separation
# ============================================================

train_orders = set(train_df["Order ID"])
test_orders = set(test_df["Order ID"])

order_overlap = train_orders.intersection(test_orders)

if len(order_overlap) != 0:
    raise ValueError(
        f"Order leakage detected: {len(order_overlap)} overlapping orders."
    )

print("\nORDER-LEVEL SPLIT VALIDATION")
print("-" * 70)
print(f"Train records: {len(train_df):,}")
print(f"Test records : {len(test_df):,}")
print(f"Train orders : {len(train_orders):,}")
print(f"Test orders  : {len(test_orders):,}")
print(f"Order overlap: {len(order_overlap)}")


# ============================================================
# 13. Create TRAIN-ONLY target threshold
# ============================================================

train_q3 = train_df["Delivery_Days"].quantile(0.75)

if not np.isfinite(train_q3):
    raise ValueError(
        "Invalid train-only Q3 threshold."
    )

train_df["Slow_Shipment"] = (
    train_df["Delivery_Days"] > train_q3
).astype(int)

test_df["Slow_Shipment"] = (
    test_df["Delivery_Days"] > train_q3
).astype(int)


# ============================================================
# 14. Validate expected split statistics
# ============================================================

if len(train_df) != 8132:
    raise ValueError(
        f"Unexpected train record count: {len(train_df)}. "
        "Expected 8,132."
    )

if len(test_df) != 2062:
    raise ValueError(
        f"Unexpected test record count: {len(test_df)}. "
        "Expected 2,062."
    )

if len(train_orders) != 6839:
    raise ValueError(
        f"Unexpected train order count: {len(train_orders)}. "
        "Expected 6,839."
    )

if len(test_orders) != 1710:
    raise ValueError(
        f"Unexpected test order count: {len(test_orders)}. "
        "Expected 1,710."
    )

if int(round(train_q3)) != 1638:
    raise ValueError(
        f"Unexpected train-only Q3 threshold: {train_q3}. "
        "Expected 1,638."
    )


# ============================================================
# 15. Define EXACT model input features
# ============================================================

feature_columns = [
    "Product Name",
    "Division",
    "Country/Region",
    "State/Province",
    "Region",
    "Factory",
    "Ship Mode",
    "Units",
    "Sales",
    "Cost",
    "Gross Profit",
    "Order_Year",
    "Order_Month",
    "Order_Quarter",
    "Order_DayOfWeek"
]


# ============================================================
# 16. Validate feature columns
# ============================================================

missing_features = [
    col for col in feature_columns
    if col not in test_df.columns
]

if missing_features:
    raise ValueError(
        f"Missing model features: {missing_features}"
    )

X_test = test_df[feature_columns].copy()
y_test = test_df["Slow_Shipment"].copy()


# ============================================================
# 17. Load saved Random Forest pipeline
# ============================================================

loaded_model = joblib.load(model_path)

if not hasattr(loaded_model, "predict_proba"):
    raise TypeError(
        "Saved model does not support predict_proba()."
    )

if not hasattr(loaded_model, "predict"):
    raise TypeError(
        "Saved model does not support predict()."
    )

print("\nMODEL VALIDATION")
print("-" * 70)
print(f"Model file: {model_path}")
print(f"Model type: {type(loaded_model).__name__}")


# ============================================================
# 18. Verify test-set predictive performance
# ============================================================

test_probabilities = loaded_model.predict_proba(X_test)[:, 1]

if not np.isfinite(test_probabilities).all():
    raise ValueError(
        "Model produced invalid probability values."
    )

baseline_auc = roc_auc_score(
    y_test,
    test_probabilities
)

print(f"Test ROC-AUC: {baseline_auc:.4f}")

# Existing validated model performance should be approximately 0.9013.
if not np.isclose(
    baseline_auc,
    0.9013,
    atol=0.01
):
    raise ValueError(
        f"Unexpected test ROC-AUC: {baseline_auc:.4f}. "
        "This does not match the previously validated model."
    )


# ============================================================
# 19. Calculate TEST-SET permutation importance
# ============================================================

print("\nCalculating permutation importance...")
print("Metric: ROC-AUC")
print("Repeats: 10")
print("Random state: 42")

permutation_result = permutation_importance(
    loaded_model,
    X_test,
    y_test,
    scoring="roc_auc",
    n_repeats=10,
    random_state=42,
    n_jobs=-1
)


# ============================================================
# 20. Validate permutation output
# ============================================================

if len(permutation_result.importances_mean) != len(feature_columns):
    raise ValueError(
        "Permutation importance count does not match "
        "the number of model input features."
    )

if not np.isfinite(
    permutation_result.importances_mean
).all():
    raise ValueError(
        "Permutation importance contains invalid values."
    )

if not np.isfinite(
    permutation_result.importances_std
).all():
    raise ValueError(
        "Permutation importance standard deviation "
        "contains invalid values."
    )


# ============================================================
# 21. Create importance table
# ============================================================

permutation_df = pd.DataFrame({
    "Feature": feature_columns,
    "Importance_Mean": permutation_result.importances_mean,
    "Importance_STD": permutation_result.importances_std
})

permutation_df["Importance_Percentage"] = (
    permutation_df["Importance_Mean"].clip(lower=0)
)

positive_total = permutation_df[
    "Importance_Percentage"
].sum()

if positive_total > 0:
    permutation_df["Importance_Percentage"] = (
        permutation_df["Importance_Percentage"]
        / positive_total
    ) * 100
else:
    permutation_df["Importance_Percentage"] = 0.0

permutation_df = permutation_df.sort_values(
    "Importance_Mean",
    ascending=False
).reset_index(drop=True)


# ============================================================
# 22. Add interpretation label
# ============================================================

def importance_level(value):
    if value > 0.02:
        return "HIGH"
    elif value > 0.005:
        return "MODERATE"
    elif value > 0:
        return "LOW"
    else:
        return "NO POSITIVE TEST-SET IMPACT"


permutation_df["Importance_Level"] = (
    permutation_df["Importance_Mean"]
    .apply(importance_level)
)


# ============================================================
# 23. Save results
# ============================================================

output_path = (
    results_dir
    / "random_forest_test_permutation_importance.csv"
)

permutation_df.to_csv(
    output_path,
    index=False
)


# ============================================================
# 24. Display final results
# ============================================================

print("\n" + "=" * 70)
print("TEST-SET PERMUTATION IMPORTANCE RESULTS")
print("=" * 70)

print(
    permutation_df[
        [
            "Feature",
            "Importance_Mean",
            "Importance_STD",
            "Importance_Percentage",
            "Importance_Level"
        ]
    ]
    .to_string(index=False)
)


# ============================================================
# 25. Final validation
# ============================================================

print("\n" + "=" * 70)
print("STEP 28 VALIDATION")
print("=" * 70)

print("Raw dataset validation: PASSED")
print("Factory mapping validation: PASSED")
print("Date parsing validation: PASSED")
print("Delivery calculation validation: PASSED")
print("Order-level split validation: PASSED")
print("Train-only target threshold validation: PASSED")
print("Feature validation: PASSED")
print("Saved model validation: PASSED")
print(f"Test ROC-AUC validation: PASSED ({baseline_auc:.4f})")
print("Permutation importance calculation: PASSED")
print("Permutation output validation: PASSED")

print(f"\nSaved file:")
print(output_path)

print("\n" + "=" * 70)
print("STATUS: STEP 28 COMPLETED SUCCESSFULLY")
print("=" * 70)
````

````python
# ============================================================
# STEP 31B — FINAL GITHUB PRE-PUSH SAFETY AUDIT
# ============================================================

from pathlib import Path
import re
import subprocess
import sys

# ------------------------------------------------------------
# 1. LOCATE PROJECT DIRECTORY
# ------------------------------------------------------------

cwd = Path.cwd().resolve()

possible_dirs = [
    cwd,
    *cwd.parents
]

project_dir = None

for p in possible_dirs:
    if (
        (p / "data" / "Nassau Candy Distributor.csv").exists()
        and (p / "models").exists()
        and (p / "results").exists()
        and (p / "notebooks").exists()
    ):
        project_dir = p
        break

if project_dir is None:
    raise FileNotFoundError(
        "Project directory could not be located automatically."
    )

print("=" * 70)
print("STEP 31B — FINAL GITHUB PRE-PUSH SAFETY AUDIT")
print("=" * 70)
print("Project directory:")
print(project_dir)

# ------------------------------------------------------------
# 2. REQUIRED FILES
# ------------------------------------------------------------

required_files = [
    "README.md",
    "requirements.txt",
    ".gitignore",

    "data/Nassau Candy Distributor.csv",

    "models/nassau_slow_shipment_model.joblib",
    "models/nassau_slow_shipment_model_metadata.json",

    "notebooks/01_data_understanding.ipynb",

    "results/factory_reallocation_scenarios.csv",
    "results/factory_reallocation_recommendations.csv",
    "results/final_factory_reallocation_recommendations.csv",
    "results/random_forest_feature_importance.csv",
    "results/random_forest_transformed_feature_importance.csv",
    "results/random_forest_test_permutation_importance.csv",
    "results/final_business_impact_summary.csv",
    "results/final_business_priority_table.csv",
]

print("\n" + "=" * 70)
print("REQUIRED FILE CHECK")
print("=" * 70)

missing_files = []

for relative_path in required_files:
    path = project_dir / relative_path

    if path.exists() and path.is_file():
        print(f"[PASSED] {relative_path}")
    else:
        print(f"[MISSING] {relative_path}")
        missing_files.append(relative_path)

if missing_files:
    raise FileNotFoundError(
        "Required project files are missing:\n"
        + "\n".join(missing_files)
    )

print("Required file check: PASSED")

# ------------------------------------------------------------
# 3. FILE SIZE AUDIT
# GitHub hard limit = 100 MiB
# Warning threshold = 50 MiB
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FILE SIZE AUDIT")
print("=" * 70)

GITHUB_HARD_LIMIT = 100 * 1024 * 1024
WARNING_LIMIT = 50 * 1024 * 1024

large_files = []
warning_files = []

for relative_path in required_files:
    path = project_dir / relative_path
    size_bytes = path.stat().st_size
    size_mb = size_bytes / (1024 * 1024)

    if size_bytes > GITHUB_HARD_LIMIT:
        status = "BLOCKED"
        large_files.append(relative_path)

    elif size_bytes > WARNING_LIMIT:
        status = "WARNING"
        warning_files.append(relative_path)

    else:
        status = "OK"

    print(f"{relative_path:<65} {size_mb:>8.2f} MB  {status}")

if large_files:
    raise ValueError(
        "One or more required files exceed GitHub's 100 MiB limit:\n"
        + "\n".join(large_files)
    )

if warning_files:
    print("\nWarning: files above 50 MiB:")
    for item in warning_files:
        print(" -", item)

print("\nFile size audit: PASSED")

# ------------------------------------------------------------
# 4. .GITIGNORE AUDIT
# ------------------------------------------------------------

print("\n" + "=" * 70)
print(".GITIGNORE AUDIT")
print("=" * 70)

gitignore_path = project_dir / ".gitignore"

gitignore_text = gitignore_path.read_text(
    encoding="utf-8",
    errors="replace"
)

required_ignore_patterns = [
    ".venv/",
    "__pycache__/",
    ".ipynb_checkpoints/",
    ".env",
    ".env.*",
]

missing_ignore_patterns = []

for pattern in required_ignore_patterns:
    if pattern not in gitignore_text:
        missing_ignore_patterns.append(pattern)

    else:
        print(f"[PASSED] {pattern}")

if missing_ignore_patterns:
    raise ValueError(
        "Required .gitignore patterns are missing:\n"
        + "\n".join(missing_ignore_patterns)
    )

print(".gitignore audit: PASSED")

# ------------------------------------------------------------
# 5. DEFINE FILES THAT ARE SAFE TO SCAN
#
# IMPORTANT:
# Do NOT recursively scan .venv or installed packages.
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("CREDENTIAL SAFETY SCAN")
print("=" * 70)

scan_extensions = {
    ".py",
    ".ipynb",
    ".json",
    ".txt",
    ".md",
    ".yml",
    ".yaml",
    ".toml",
    ".ini",
    ".cfg",
    ".conf",
}

excluded_directories = {
    ".venv",
    "venv",
    "env",
    "ENV",
    ".git",
    "__pycache__",
    ".ipynb_checkpoints",
    ".pytest_cache",
    ".mypy_cache",
    "node_modules",
    "dist",
    "build",
}

files_to_scan = []

for path in project_dir.rglob("*"):

    if not path.is_file():
        continue

    # Ignore excluded directories
    if any(part in excluded_directories for part in path.parts):
        continue

    # Ignore very large/binary files
    if path.suffix.lower() not in scan_extensions:
        continue

    files_to_scan.append(path)

print("Files selected for safety scan:", len(files_to_scan))

# ------------------------------------------------------------
# 6. HIGH-CONFIDENCE SECRET PATTERNS
#
# Generic words such as "api_key" alone are NOT considered
# a credential.
# ------------------------------------------------------------

secret_patterns = [

    # OpenAI-style secret key
    (
        "OpenAI-style API key",
        re.compile(r"\bsk-[A-Za-z0-9_-]{20,}\b")
    ),

    # GitHub personal access token
    (
        "GitHub token",
        re.compile(r"\bgh[pousr]_[A-Za-z0-9_]{20,}\b")
    ),

    # AWS access key
    (
        "AWS access key",
        re.compile(r"\bAKIA[0-9A-Z]{16}\b")
    ),

    # Private key block
    (
        "Private key block",
        re.compile(r"-----BEGIN [A-Z ]*PRIVATE KEY-----")
    ),

    # Bearer token with a substantial token value
    (
        "Bearer token",
        re.compile(
            r"(?i)\bbearer\s+[A-Za-z0-9._-]{30,}\b"
        )
    ),

    # URL containing username/password credentials
    (
        "Credential-containing URL",
        re.compile(
            r"https?://[^/\s:@]+:[^/\s@]+@"
        )
    ),

    # High-confidence assignment pattern.
    # Placeholder values are explicitly excluded.
    (
        "Hard-coded secret assignment",
        re.compile(
            r"""(?ix)
            \b(api[_-]?key|secret[_-]?key|access[_-]?token|
               auth[_-]?token|password)
            \b
            \s*[:=]\s*
            ["']
            (?!your[_-]?api[_-]?key
               |your[_-]?password
               |your[_-]?token
               |change[_-]?me
               |replace[_-]?me
               |example
               |placeholder
               |none
               |null
               |xxx+
               |\*+)
            [^"']{12,}
            ["']
            """
        )
    ),
]

secret_hits = []

# ------------------------------------------------------------
# 7. SCAN ONLY PROJECT-RELEVANT TEXT FILES
# ------------------------------------------------------------

for path in files_to_scan:

    try:
        text = path.read_text(
            encoding="utf-8",
            errors="replace"
        )
    except Exception:
        continue

    lines = text.splitlines()

    for line_number, line in enumerate(lines, start=1):

        for description, pattern in secret_patterns:

            if pattern.search(line):

                relative_path = path.relative_to(project_dir)

                secret_hits.append(
                    (
                        str(relative_path),
                        line_number,
                        description,
                        line.strip()
                    )
                )

# ------------------------------------------------------------
# 8. REPORT RESULT
# ------------------------------------------------------------

if secret_hits:

    print("\nPotential high-confidence credential content found:")

    for file_name, line_number, description, line in secret_hits:
        print("\nFile:", file_name)
        print("Line:", line_number)
        print("Type:", description)

        # Do NOT print the full possible secret.
        # Mask suspicious-looking long values.
        masked_line = re.sub(
            r'([A-Za-z0-9_\-]{8})[A-Za-z0-9_\-]{8,}',
            r'\1********',
            line
        )

        print("Content:", masked_line)

    raise ValueError(
        "\nCredential safety scan FAILED. "
        "Review the listed project files before pushing."
    )

else:
    print("High-confidence credential scan: PASSED")
    print("No high-confidence API key/token/private-key pattern detected.")

# ------------------------------------------------------------
# 9. CHECK GIT INSTALLATION
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("GIT ENVIRONMENT CHECK")
print("=" * 70)

try:
    git_version = subprocess.check_output(
        ["git", "--version"],
        text=True,
        stderr=subprocess.STDOUT
    ).strip()

    print("Git:", git_version)
    git_installed = True

except Exception:
    print("Git: NOT FOUND")
    git_installed = False

# ------------------------------------------------------------
# 10. CHECK WHETHER REPOSITORY ALREADY EXISTS
# ------------------------------------------------------------

git_dir = project_dir / ".git"

if git_dir.exists() and git_dir.is_dir():

    print("Git repository: EXISTS")

    try:
        remote_output = subprocess.check_output(
            ["git", "remote", "-v"],
            cwd=project_dir,
            text=True,
            stderr=subprocess.STDOUT
        ).strip()

        if remote_output:
            print("Git remote:")
            print(remote_output)
        else:
            print("Git remote: NOT CONFIGURED")

    except Exception as e:
        print("Git remote check could not be completed:", e)

else:
    print("Git repository: NOT INITIALIZED")
    print("This is OK for the current pre-push audit.")

# ------------------------------------------------------------
# 11. REQUIRED FILE TOTAL SIZE
# ------------------------------------------------------------

total_bytes = 0

for relative_path in required_files:
    total_bytes += (project_dir / relative_path).stat().st_size

total_mb = total_bytes / (1024 * 1024)

print("\n" + "=" * 70)
print("PROJECT SIZE SUMMARY")
print("=" * 70)

print(f"Required project files: {len(required_files)}")
print(f"Combined size: {total_mb:.2f} MB")

# ------------------------------------------------------------
# 12. FINAL STATUS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FINAL PRE-PUSH DECISION")
print("=" * 70)

print("Required files: PASSED")
print("File size limits: PASSED")
print(".gitignore: PASSED")
print("Credential safety scan: PASSED")

if git_installed:
    print("Git installation: PASSED")
else:
    print("Git installation: NOT AVAILABLE")

print("\nIMPORTANT:")
print("No git add was performed.")
print("No commit was created.")
print("No push was performed.")

print("\nSTATUS: STEP 31B COMPLETED SUCCESSFULLY")
````
